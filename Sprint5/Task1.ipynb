<h1 style="color:DodgerBlue">HashSet<T></h1>



`HashSet<T>` в C# — это коллекция, которая хранит уникальные элементы без определенного порядка. Она полезна, когда нужно быстро проверять наличие элемента или добавлять только уникальные элементы. Давайте рассмотрим, как с ней работать на примере классов `Transport`, `Car`, и `Bike`.

### Создание классов Transport, Car, и Bike

Сначала определим базовый класс `Transport` и его наследники `Car` и `Bike`:

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public override bool Equals(object obj)
    {
        // Простая проверка равенства базируется на модели и производителе
        if (obj is Transport transport)
        {
            return Model == transport.Model && Manufacturer == transport.Manufacturer;
        }
        return false;
    }

    public override int GetHashCode()
    {
        // Создание хеш-кода на основе модели и производителя
        return HashCode.Combine(Model, Manufacturer);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Number of Doors: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }
}
```

### Работа с HashSet<T>

Теперь рассмотрим, как использовать `HashSet<T>` для хранения объектов `Transport`.

```csharp
using System;
using System.Collections.Generic;


// Создаем HashSet для уникальных объектов транспортных средств
HashSet<Transport> transportSet = new HashSet<Transport>();

// Добавляем объекты Car и Bike в HashSet
var car1 = new Car("Model S", "Tesla", 4);
var car2 = new Car("Model 3", "Tesla", 4);
 var bike1 = new Bike("Mountain Explorer", "Giant", true);

transportSet.Add(car1);
transportSet.Add(car2);
transportSet.Add(bike1);

// Попробуем добавить повторяющийся объект
var duplicateCar = new Car("Model S", "Tesla", 4);
bool isAdded = transportSet.Add(duplicateCar);

Console.WriteLine($"Was duplicate car added? {isAdded}"); // false

// Перебираем и выводим информацию обо всех уникальных транспортных средствах
foreach (var transport in transportSet)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

```

### Объяснения и комментарии:

- **Уникальность объектов**: `HashSet<T>` не допускает дублирования. Если вы пытаетесь добавить элемент, который уже присутствует (определяется методом `Equals` и `GetHashCode`), он не добавляется.

- **Методы `Equals` и `GetHashCode`**: Эти методы переопределены для корректной работы с `HashSet`. `Equals` определяет, когда два объекта считаются равными, а `GetHashCode` возвращает хеш-код, используемый для быстрого поиска элемента.

- **Быстрая проверка на наличие**: Благодаря хешированию, `HashSet` позволяет быстро проверять, существует ли элемент в коллекции (`O(1)` — амортизированное время).

- **Отсутствие порядка**: Элементы в `HashSet` не имеют определенного порядка, в отличие от `List<T>`.

`HashSet<T>` идеально подходит для сценариев, где требуется исключительно уникальность элементов и быстрое выполнение операций поиска. 

<h4 style="color:DodgerBlue">Для проверки напишите пример кода на основе классов `Transport`, `Car`, и `Bike` ниже в блоке с применением  HashSet</h4>

````
using System;
using System.Collections.Generic;

// Базовый класс Transport
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public override bool Equals(object obj)
    {
        if (obj is Transport transport)
        {
            return Model == transport.Model && Manufacturer == transport.Manufacturer;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, Manufacturer);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }
}

// Класс Car
public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }
}

// Класс Bike
public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет носитель: {HasCarrier}");
    }
}

Console.WriteLine("=== ПРОВЕРКА РАБОТЫ С HASHSET ===\n");

// 1. СОЗДАНИЕ HASHSET
Console.WriteLine("1. СОЗДАНИЕ И ЗАПОЛНЕНИЕ HASHSET:\n");
HashSet<Transport> transportSet = new HashSet<Transport>();

// Создаем объекты
var car1 = new Car("Model S", "Tesla", 4);
var car2 = new Car("Model 3", "Tesla", 4);
var car3 = new Car("Camry", "Toyota", 4);
var bike1 = new Bike("Mountain Explorer", "Giant", true);
var bike2 = new Bike("Road Master", "Trek", false);

// Добавляем объекты
transportSet.Add(car1);
transportSet.Add(car2);
transportSet.Add(car3);
transportSet.Add(bike1);
transportSet.Add(bike2);

Console.WriteLine($"Количество элементов в HashSet: {transportSet.Count}");

// 2. ПРОВЕРКА УНИКАЛЬНОСТИ
Console.WriteLine("\n2. ПРОВЕРКА УНИКАЛЬНОСТИ ЭЛЕМЕНТОВ:\n");

// Попробуем добавить повторяющийся объект
var duplicateCar = new Car("Model S", "Tesla", 4);
bool isAdded = transportSet.Add(duplicateCar);

Console.WriteLine($"Был ли добавлен дубликат (Model S, Tesla)? {isAdded}");
Console.WriteLine($"Теперь количество элементов: {transportSet.Count}");

// Попробуем добавить еще один дубликат
var anotherDuplicate = new Car("Camry", "Toyota", 4);
isAdded = transportSet.Add(anotherDuplicate);
Console.WriteLine($"Был ли добавлен дубликат (Camry, Toyota)? {isAdded}");
Console.WriteLine($"Теперь количество элементов: {transportSet.Count}");

// 3. ВЫВОД ВСЕХ УНИКАЛЬНЫХ ЭЛЕМЕНТОВ
Console.WriteLine("\n3. ВСЕ УНИКАЛЬНЫЕ ТРАНСПОРТНЫЕ СРЕДСТВА:\n");
foreach (var transport in transportSet)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

// 4. МЕТОДЫ HASHSET
Console.WriteLine("4. ОСНОВНЫЕ МЕТОДЫ HASHSET:\n");

// Проверка наличия элемента
var carToCheck = new Car("Model 3", "Tesla", 4);
bool containsCar = transportSet.Contains(carToCheck);
Console.WriteLine($"Содержит ли HashSet машину Model 3? {containsCar}");

// Проверка несуществующего элемента
var nonExistentCar = new Car("Focus", "Ford", 4);
containsCar = transportSet.Contains(nonExistentCar);
Console.WriteLine($"Содержит ли HashSet машину Focus? {containsCar}");

// Удаление элемента
bool removed = transportSet.Remove(bike1);
Console.WriteLine($"Удален ли Mountain Explorer? {removed}");
Console.WriteLine($"Количество элементов после удаления: {transportSet.Count}");

// 5. СРАВНЕНИЕ С LIST
Console.WriteLine("\n5. СРАВНЕНИЕ HASHSET И LIST:\n");

// Создаем List с дубликатами
List<Transport> transportList = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Car("Model S", "Tesla", 4), // Дубликат
    new Car("Model 3", "Tesla", 4),
    new Bike("Mountain Explorer", "Giant", true),
    new Bike("Mountain Explorer", "Giant", true) // Дубликат
};

Console.WriteLine($"List содержит {transportList.Count} элементов (с дубликатами)");

// Создаем HashSet из List (автоматически убирает дубликаты)
HashSet<Transport> uniqueSet = new HashSet<Transport>(transportList);
Console.WriteLine($"HashSet содержит {uniqueSet.Count} уникальных элементов");

// 6. ОПЕРАЦИИ С HASHSET
Console.WriteLine("\n6. ОПЕРАЦИИ НАД HASHSET:\n");

// Создаем еще один HashSet
HashSet<Transport> anotherSet = new HashSet<Transport>
{
    new Car("Camry", "Toyota", 4),
    new Bike("Road Master", "Trek", false),
    new Car("Civic", "Honda", 4)
};

// Объединение множеств (UnionWith)
HashSet<Transport> unionSet = new HashSet<Transport>(transportSet);
unionSet.UnionWith(anotherSet);
Console.WriteLine($"Объединение: {unionSet.Count} элементов");

// Пересечение множеств (IntersectWith)
HashSet<Transport> intersectSet = new HashSet<Transport>(transportSet);
intersectSet.IntersectWith(anotherSet);
Console.WriteLine($"Пересечение: {intersectSet.Count} общих элементов");

// Разность множеств (ExceptWith)
HashSet<Transport> exceptSet = new HashSet<Transport>(transportSet);
exceptSet.ExceptWith(anotherSet);
Console.WriteLine($"Разность: {exceptSet.Count} элементов только в первом множестве");

// 7. ИТЕРАЦИЯ ПО HASHSET
Console.WriteLine("\n7. РАЗЛИЧНЫЕ СПОСОБЫ ИТЕРАЦИИ:\n");

Console.WriteLine("Итерация с помощью foreach:");
foreach (var transport in transportSet)
{
    Console.WriteLine($"- {transport.Model} ({transport.Manufacturer})");
}

// Преобразование в массив
Console.WriteLine("\nПреобразование в массив:");
Transport[] transportArray = new Transport[transportSet.Count];
transportSet.CopyTo(transportArray);

for (int i = 0; i < transportArray.Length; i++)
{
    Console.WriteLine($"{i + 1}. {transportArray[i].Model}");
}
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте при помощи  HashSet. 

````
using System;
using System.Collections.Generic;

// Базовый класс Animal с переопределенными Equals и GetHashCode
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Species { get; set; }

    public Animal(string name, int age, string species)
    {
        Name = name;
        Age = age;
        Species = species;
    }

    public override bool Equals(object obj)
    {
        if (obj is Animal animal)
        {
            return Name == animal.Name && 
                   Age == animal.Age && 
                   Species == animal.Species;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Age, Species);
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Животное: {Name}, Вид: {Species}, Возраст: {Age} лет");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, string breed)
        : base(name, age, "Собака")
    {
        Breed = breed;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }

    // Переопределяем Equals для учета породы
    public override bool Equals(object obj)
    {
        if (obj is Dog dog)
        {
            return base.Equals(dog) && Breed == dog.Breed;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), Breed);
    }
}

// Производный класс Cat
public class Cat : Animal
{
    public string Color { get; set; }

    public Cat(string name, int age, string color)
        : base(name, age, "Кот")
    {
        Color = color;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Цвет: {Color}");
    }

    // Переопределяем Equals для учета цвета
    public override bool Equals(object obj)
    {
        if (obj is Cat cat)
        {
            return base.Equals(cat) && Color == cat.Color;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), Color);
    }
}

// Производный класс Bird
public class Bird : Animal
{
    public bool CanFly { get; set; }

    public Bird(string name, int age, bool canFly)
        : base(name, age, "Птица")
    {
        CanFly = canFly;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Умеет летать: {CanFly}");
    }

    // Переопределяем Equals для учета возможности полета
    public override bool Equals(object obj)
    {
        if (obj is Bird bird)
        {
            return base.Equals(bird) && CanFly == bird.CanFly;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), CanFly);
    }
}

// Производный класс Fish
public class Fish : Animal
{
    public string WaterType { get; set; }

    public Fish(string name, int age, string waterType)
        : base(name, age, "Рыба")
    {
        WaterType = waterType;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} издает булькающие звуки");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}");
    }

    // Переопределяем Equals для учета типа воды
    public override bool Equals(object obj)
    {
        if (obj is Fish fish)
        {
            return base.Equals(fish) && WaterType == fish.WaterType;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), WaterType);
    }
}

Console.WriteLine("=== ДЕМОНСТРАЦИЯ РАБОТЫ С HASHSET<ANIMAL> ===\n");

Console.WriteLine("1. СОЗДАНИЕ HASHSET И ДОБАВЛЕНИЕ ЖИВОТНЫХ:\n");
HashSet<Animal> animalSet = new HashSet<Animal>();

// Создаем животных
var dog1 = new Dog("Шарик", 3, "Овчарка");
var dog2 = new Dog("Рекс", 4, "Доберман");
var cat1 = new Cat("Барсик", 2, "рыжий");
var cat2 = new Cat("Мурка", 3, "серая");
var bird1 = new Bird("Кеша", 1, true);
var bird2 = new Bird("Гоша", 2, false);
var fish1 = new Fish("Немо", 1, "морская");
var fish2 = new Fish("Золотистый", 2, "пресная");

Console.WriteLine("Добавление животных в HashSet:");
Console.WriteLine($"Добавлен Шарик: {animalSet.Add(dog1)}");
Console.WriteLine($"Добавлен Рекс: {animalSet.Add(dog2)}");
Console.WriteLine($"Добавлен Барсик: {animalSet.Add(cat1)}");
Console.WriteLine($"Добавлен Мурка: {animalSet.Add(cat2)}");
Console.WriteLine($"Добавлен Кеша: {animalSet.Add(bird1)}");
Console.WriteLine($"Добавлен Гоша: {animalSet.Add(bird2)}");
Console.WriteLine($"Добавлен Немо: {animalSet.Add(fish1)}");
Console.WriteLine($"Добавлен Золотистый: {animalSet.Add(fish2)}");

Console.WriteLine($"\nВсего уникальных животных: {animalSet.Count}");

Console.WriteLine("\n2. ПРОВЕРКА УНИКАЛЬНОСТИ ЖИВОТНЫХ:\n");

// Попробуем добавить дубликаты
var duplicateDog = new Dog("Шарик", 3, "Овчарка");
var duplicateCat = new Cat("Барсик", 2, "рыжий");

Console.WriteLine($"Пытаемся добавить дубликат Шарика: {animalSet.Add(duplicateDog)}");
Console.WriteLine($"Пытаемся добавить дубликат Барсика: {animalSet.Add(duplicateCat)}");
Console.WriteLine($"Количество после попытки добавить дубликаты: {animalSet.Count}");

// Попробуем добавить похожее, но не идентичное животное
var similarDog = new Dog("Шарик", 3, "Дворняжка"); // Другая порода
Console.WriteLine($"\nПытаемся добавить Шарика другой породы: {animalSet.Add(similarDog)}");
Console.WriteLine($"Количество после добавления похожего животного: {animalSet.Count}");


Console.WriteLine("\n3. ВСЕ УНИКАЛЬНЫЕ ЖИВОТНЫЕ В HASHSET:\n");
int counter = 1;
foreach (var animal in animalSet)
{
    Console.WriteLine($"{counter}. {animal.GetType().Name}:");
    animal.DisplayInfo();
    animal.MakeSound();
    Console.WriteLine();
    counter++;
}

Console.WriteLine("4. ОСНОВНЫЕ МЕТОДЫ HASHSET ДЛЯ РАБОТЫ С ЖИВОТНЫМИ:\n");

// Проверка наличия животного
var checkDog = new Dog("Рекс", 4, "Доберман");
var checkCat = new Cat("Васька", 5, "черный");

Console.WriteLine($"Содержит ли HashSet собаку Рекс? {animalSet.Contains(checkDog)}");
Console.WriteLine($"Содержит ли HashSet кота Васька? {animalSet.Contains(checkCat)}");

// Удаление животного
bool removed = animalSet.Remove(cat1);
Console.WriteLine($"\nУдален ли кот Барсик? {removed}");
Console.WriteLine($"Количество животных после удаления: {animalSet.Count}");

// Очистка части HashSet
Console.WriteLine("\nУдаляем всех птиц:");
int birdsRemoved = animalSet.RemoveWhere(animal => animal is Bird);
Console.WriteLine($"Удалено птиц: {birdsRemoved}");
Console.WriteLine($"Осталось животных: {animalSet.Count}");

// 5. ОПЕРАЦИИ НАД HASHSET
Console.WriteLine("\n5. ОПЕРАЦИИ НАД МНОЖЕСТВАМИ ЖИВОТНЫХ:\n");

// Создаем еще один HashSet с животными
HashSet<Animal> anotherAnimalSet = new HashSet<Animal>
{
    new Dog("Бобик", 2, "Дворняжка"),
    new Cat("Васька", 5, "черный"),
    new Fish("Немо", 1, "морская"), // Дубликат
    new Bird("Соловей", 1, true)
};

Console.WriteLine("Второй HashSet содержит:");
foreach (var animal in anotherAnimalSet)
{
    Console.WriteLine($"- {animal.Name}");
}

// Объединение множеств
HashSet<Animal> unionSet = new HashSet<Animal>(animalSet);
unionSet.UnionWith(anotherAnimalSet);
Console.WriteLine($"\nОбъединение двух HashSet: {unionSet.Count} уникальных животных");

// Пересечение множеств
HashSet<Animal> intersectSet = new HashSet<Animal>(animalSet);
intersectSet.IntersectWith(anotherAnimalSet);
Console.WriteLine($"Пересечение (общие животные): {intersectSet.Count}");

if (intersectSet.Count > 0)
{
    Console.WriteLine("Общие животные:");
    foreach (var animal in intersectSet)
    {
        Console.WriteLine($"- {animal.Name}");
    }
}

// Разность множеств
HashSet<Animal> exceptSet = new HashSet<Animal>(animalSet);
exceptSet.ExceptWith(anotherAnimalSet);
Console.WriteLine($"\nРазность (только в первом HashSet): {exceptSet.Count}");
foreach (var animal in exceptSet)
{
    Console.WriteLine($"- {animal.Name}");
}

Console.WriteLine("\n6. ГРУППИРОВКА ЖИВОТНЫХ ПО ТИПУ:\n");

// Создаем HashSet для каждого типа животных
HashSet<Dog> dogs = new HashSet<Dog>();
HashSet<Cat> cats = new HashSet<Cat>();
HashSet<Bird> birds = new HashSet<Bird>();
HashSet<Fish> fishes = new HashSet<Fish>();

foreach (var animal in unionSet)
{
    if (animal is Dog dog) dogs.Add(dog);
    else if (animal is Cat cat) cats.Add(cat);
    else if (animal is Bird bird) birds.Add(bird);
    else if (animal is Fish fish) fishes.Add(fish);
}

Console.WriteLine($"Собак: {dogs.Count}");
Console.WriteLine($"Котов: {cats.Count}");
Console.WriteLine($"Птиц: {birds.Count}");
Console.WriteLine($"Рыб: {fishes.Count}");

Console.WriteLine("\n7. УДАЛЕНИЕ ДУБЛИКАТОВ ИЗ LIST С ПОМОЩЬЮ HASHSET:\n");

// Создаем List с дубликатами животных
List<Animal> animalListWithDuplicates = new List<Animal>
{
    new Dog("Шарик", 3, "Овчарка"),
    new Dog("Шарик", 3, "Овчарка"), // Дубликат
    new Cat("Барсик", 2, "рыжий"),
    new Cat("Барсик", 2, "рыжий"), // Дубликат
    new Dog("Рекс", 4, "Доберман"),
    new Bird("Кеша", 1, true),
    new Bird("Кеша", 1, true) // Дубликат
};

Console.WriteLine($"List содержит {animalListWithDuplicates.Count} элементов (с дубликатами)");

// Используем HashSet для удаления дубликатов
HashSet<Animal> uniqueAnimalsFromList = new HashSet<Animal>(animalListWithDuplicates);
Console.WriteLine($"После удаления дубликатов: {uniqueAnimalsFromList.Count} уникальных животных");

// Преобразуем обратно в List без дубликатов
List<Animal> uniqueAnimalList = new List<Animal>(uniqueAnimalsFromList);
Console.WriteLine($"Уникальный List содержит: {uniqueAnimalList.Count} животных");

Console.WriteLine("\n8. ДЕМОНСТРАЦИЯ ПРОИЗВОДИТЕЛЬНОСТИ HASHSET:\n");

// HashSet для быстрого поиска
HashSet<string> animalNames = new HashSet<string>();
foreach (var animal in animalSet)
{
    animalNames.Add(animal.Name);
}

Console.WriteLine("Быстрая проверка имен животных:");
string[] namesToCheck = { "Шарик", "Рекс", "Васька", "Немо" };
foreach (var name in namesToCheck)
{
    bool exists = animalNames.Contains(name);
    Console.WriteLine($"Имя '{name}' существует: {exists}");
}

Console.WriteLine("\n9. СТАТИСТИКА ПО HASHSET ЖИВОТНЫХ:\n");
Console.WriteLine($"Всего уникальных животных: {unionSet.Count}");
Console.WriteLine($"Средний возраст: {unionSet.Average(a => a.Age):F1} лет");
Console.WriteLine($"Самый старый: {unionSet.OrderByDescending(a => a.Age).First().Name} ({unionSet.Max(a => a.Age)} лет)");
Console.WriteLine($"Самый молодой: {unionSet.OrderBy(a => a.Age).First().Name} ({unionSet.Min(a => a.Age)} лет)");

// Подсчет по видам
Console.WriteLine("\nРаспределение по видам:");
Console.WriteLine($"Собаки: {unionSet.Count(a => a is Dog)}");
Console.WriteLine($"Коты: {unionSet.Count(a => a is Cat)}");
Console.WriteLine($"Птицы: {unionSet.Count(a => a is Bird)}");
Console.WriteLine($"Рыбы: {unionSet.Count(a => a is Fish)}");
````
