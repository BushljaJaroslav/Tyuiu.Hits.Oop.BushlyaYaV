<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport

    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
```
```csharp
    }
}
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

````
using System;
using System.Collections.Generic;
using System.Linq;

// ДЕЛЕГАТЫ
public delegate void DisplayDelegate();
public delegate bool FilterDelegate(Transport transport);
public delegate void TransportEventHandler(string message);

// БАЗОВЫЙ КЛАСС TRANSPORT С СОБЫТИЕМ
public class Transport
{
    public string Model { get; private set; }
    public string Manufacturer { get; private set; }

    // СОБЫТИЕ
    public event TransportEventHandler TransportUpdated;

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }

    protected virtual void OnTransportUpdated(string message)
    {
        TransportUpdated?.Invoke(message);
    }

    public void UpdateModel(string newModel)
    {
        string oldModel = Model;
        Model = newModel;
        // ВЫЗЫВАЕМ СОБЫТИЕ ЧЕРЕЗ ЗАЩИЩЕННЫЙ МЕТОД
        OnTransportUpdated($"Model updated from '{oldModel}' to '{newModel}'");
    }

    public void UpdateManufacturer(string newManufacturer)
    {
        string oldManufacturer = Manufacturer;
        Manufacturer = newManufacturer;
        OnTransportUpdated($"Manufacturer updated from '{oldManufacturer}' to '{newManufacturer}'");
    }
}

// КЛАСС CAR
public class Car : Transport
{
    public int NumberOfDoors { get; private set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Number of Doors: {NumberOfDoors}");
    }

    protected override void OnTransportUpdated(string message)
    {
        base.OnTransportUpdated(message);
    }

    public void UpdateDoors(int newNumberOfDoors)
    {
        int oldDoors = NumberOfDoors;
        NumberOfDoors = newNumberOfDoors;
        OnTransportUpdated($"Number of doors updated from {oldDoors} to {newNumberOfDoors}");
    }
}

// КЛАСС BIKE
public class Bike : Transport
{
    public bool HasCarrier { get; private set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }

    public void UpdateCarrier(bool newHasCarrier)
    {
        bool oldCarrier = HasCarrier;
        HasCarrier = newHasCarrier;
        OnTransportUpdated($"Carrier updated from '{oldCarrier}' to '{newHasCarrier}'");
    }
}

// ПРОВЕРКА РАБОТЫ
Console.WriteLine("=== ПРОВЕРКА ДЕЛЕГАТОВ, ЛЯМБДА-ВЫРАЖЕНИЙ И СОБЫТИЙ ===\n");

// 1. ДЕЛЕГАТЫ
Console.WriteLine("1. РАБОТА С ДЕЛЕГАТАМИ:\n");

// Создаем объекты
Car car = new Car("Model S", "Tesla", 4);
Bike bike = new Bike("Speedster", "Giant", true);

// Присваиваем делегату метод DisplayInfo
DisplayDelegate displayCar = car.DisplayInfo;
DisplayDelegate displayBike = bike.DisplayInfo;

// Вызов методов через делегаты
Console.WriteLine("Вызов методов через делегаты:");
displayCar();
Console.WriteLine();
displayBike();

// Мультикаст делегат
Console.WriteLine("\nМультикаст делегат:");
DisplayDelegate multiDisplay = car.DisplayInfo;
multiDisplay += bike.DisplayInfo;
multiDisplay += () => Console.WriteLine("Дополнительное сообщение от лямбды!");
multiDisplay();

// 2. ЛЯМБДА-ВЫРАЖЕНИЯ
Console.WriteLine("\n2. РАБОТА С ЛЯМБДА-ВЫРАЖЕНИЯМИ:\n");

List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
    new Car("Camry", "Toyota", 4),
    new Bike("Mountain", "Trek", false)
};

// Используем лямбда-выражение для фильтрации
Console.WriteLine("Все автомобили Tesla (лямбда-выражение):");
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");
foreach (var c in teslaCars)
{
    c.DisplayInfo();
}

// Делегат для фильтрации с лямбдой
Console.WriteLine("\nТранспорт с более чем 3 дверями (делегат + лямбда):");
FilterDelegate filterByDoors = t => t is Car carObj && carObj.NumberOfDoors > 3;
var filteredTransports = transports.Where(t => filterByDoors(t));
foreach (var t in filteredTransports)
{
    t.DisplayInfo();
}

// Сортировка с лямбдой
Console.WriteLine("\nСортировка по производителю (лямбда):");
var sorted = transports.OrderBy(t => t.Manufacturer).ThenBy(t => t.Model);
foreach (var t in sorted)
{
    Console.WriteLine($"{t.Manufacturer} - {t.Model}");
}

// 3. СОБЫТИЯ
Console.WriteLine("\n3. РАБОТА С СОБЫТИЯМИ:\n");

// Метод-обработчик события
void OnTransportUpdated(string message)
{
    Console.WriteLine($"[СОБЫТИЕ] {DateTime.Now:HH:mm:ss}: {message}");
}

// Подписываемся на события
car.TransportUpdated += OnTransportUpdated;
bike.TransportUpdated += OnTransportUpdated;

// Добавляем дополнительные обработчики через лямбды
car.TransportUpdated += (msg) => Console.WriteLine($"[ЛЯМБДА-ОБРАБОТЧИК] Автомобиль: {msg}");
bike.TransportUpdated += (msg) => Console.WriteLine($"[ЛЯМБДА-ОБРАБОТЧИК] Велосипед: {msg}");

// Вызываем события через обновление объектов
Console.WriteLine("Обновляем автомобиль:");
car.UpdateModel("Model X");
car.UpdateDoors(5);

Console.WriteLine("\nОбновляем велосипед:");
bike.UpdateModel("Speedster Pro");
bike.UpdateCarrier(false);

// 4. КОМБИНАЦИЯ ВСЕГО
Console.WriteLine("\n4. КОМБИНАЦИЯ ДЕЛЕГАТОВ, ЛЯМБД И СОБЫТИЙ:\n");

// Создаем коллекцию действий с использованием делегатов
List<Action> actions = new List<Action>
{
    () => car.DisplayInfo(),
    () => bike.DisplayInfo(),
    () => Console.WriteLine("Произвольное действие 1"),
    () => Console.WriteLine("Произвольное действие 2")
};

Console.WriteLine("Выполняем коллекцию действий:");
foreach (var action in actions)
{
    action();
}

// Делегат Func с лямбдой
Func<Transport, string> getDescription = t => 
    $"{t.GetType().Name}: {t.Manufacturer} {t.Model}";

Console.WriteLine("\nОписания транспорта (Func делегат):");
foreach (var transport in transports)
{
    Console.WriteLine(getDescription(transport));
}

// 5. УСЛОВНЫЕ ДЕЛЕГАТЫ
Console.WriteLine("\n5. УСЛОВНЫЕ ДЕЛЕГАТЫ С ЛЯМБДАМИ:\n");

// Делегат для проверки условия
Predicate<Transport> isTesla = t => t.Manufacturer == "Tesla";
Predicate<Transport> hasCarrier = t => t is Bike bikeObj && bikeObj.HasCarrier;

Console.WriteLine("Транспорт Tesla:");
var teslaTransports = transports.FindAll(new Predicate<Transport>(isTesla));
foreach (var t in teslaTransports)
{
    t.DisplayInfo();
}

Console.WriteLine("\nВелосипеды с багажником:");
var bikesWithCarrier = transports.FindAll(new Predicate<Transport>(hasCarrier));
foreach (var t in bikesWithCarrier)
{
    t.DisplayInfo();
}

// 6. ОТПИСКА ОТ СОБЫТИЙ
Console.WriteLine("\n6. ОТПИСКА ОТ СОБЫТИЙ:\n");

// Создаем отдельный метод для отписки
void SpecialHandler(string msg) => Console.WriteLine($"[СПЕЦИАЛЬНЫЙ] {msg}");

car.TransportUpdated += SpecialHandler;
Console.WriteLine("После добавления специального обработчика:");
car.UpdateModel("Model Y");

// Отписываемся
car.TransportUpdated -= SpecialHandler;
Console.WriteLine("\nПосле удаления специального обработчика:");
car.UpdateModel("Model Z");

// 7. АНОНИМНЫЕ МЕТОДЫ (ДО ЛЯМБД)
Console.WriteLine("\n7. АНОНИМНЫЕ МЕТОДЫ:\n");

// Анонимный метод (до появления лямбд)
DisplayDelegate anonymousDelegate = delegate()
{
    Console.WriteLine("Это анонимный метод!");
    Console.WriteLine("Он выполняется как обычный метод");
};

anonymousDelegate();

// Сравнение с лямбдой
DisplayDelegate lambdaDelegate = () =>
{
    Console.WriteLine("Это лямбда-выражение!");
    Console.WriteLine("Более компактный синтаксис");
};

lambdaDelegate();
````

````
using System;
using System.Collections.Generic;
using System.Linq;

// 1. Объявляем делегат
public delegate void AnimalDisplayDelegate();

// 2. Базовый класс Animal
public class Animal
{
    public string Name { get; private set; }
    public int Age { get; private set; }
    public string Species { get; protected set; }
    
    public delegate void AnimalEventHandler(string message);
    public event AnimalEventHandler AgeChanged;
    
    public Animal(string name, int age, string species)
    {
        Name = name;
        Age = age;
        Species = species;
    }
    
    public void SetAge(int newAge)
    {
        int oldAge = Age;
        Age = newAge;
        AgeChanged?.Invoke($"Возраст животного '{Name}' изменен с {oldAge} на {newAge} лет");
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Вид: {Species}, Имя: {Name}, Возраст: {Age} лет");
    }
    
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук");
    }
}

// 3. Производные классы
public class Dog : Animal
{
    public string Breed { get; set; }
    
    public Dog(string name, int age, string breed) 
        : base(name, age, "Собака")
    {
        Breed = breed;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Гав-гав!");
    }
}

public class Cat : Animal
{
    public string Color { get; set; }
    
    public Cat(string name, int age, string color) 
        : base(name, age, "Кошка")
    {
        Color = color;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Цвет: {Color}");
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Мяу-мяу!");
    }
}

public class Bird : Animal
{
    public bool CanSing { get; set; }
    
    public Bird(string name, int age, bool canSing) 
        : base(name, age, "Птица")
    {
        CanSing = canSing;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Поет: {(CanSing ? "Да" : "Нет")}");
    }
    
    public override void MakeSound()
    {
        if (CanSing)
            Console.WriteLine($"{Name} поет: Чик-чирик!");
        else
            Console.WriteLine($"{Name} издает звуки: Цвинь-цвинь!");
    }
}

// 4. Тестовый код - ВАЖНО: этот код ДОЛЖЕН выполняться!
Console.WriteLine("=== ТЕСТИРОВАНИЕ КЛАССОВ ANIMAL ===\n");

// Создаем животных
Dog dog = new Dog("Барсик", 3, "Лабрадор");
Cat cat = new Cat("Мурка", 2, "Рыжий");
Bird bird = new Bird("Кеша", 1, true);

// Тестируем базовые методы
Console.WriteLine("1. Базовые методы:");
dog.DisplayInfo();
cat.DisplayInfo();
bird.DisplayInfo();

Console.WriteLine("\n2. Звуки животных:");
dog.MakeSound();
cat.MakeSound();
bird.MakeSound();

// Тестируем делегаты
Console.WriteLine("\n3. ДЕЛЕГАТЫ:");
AnimalDisplayDelegate dogDelegate = dog.DisplayInfo;
AnimalDisplayDelegate catDelegate = cat.DisplayInfo;
AnimalDisplayDelegate birdDelegate = bird.DisplayInfo;

Console.WriteLine("Вызов через делегаты:");
dogDelegate();
catDelegate();
birdDelegate();

// Мультикаст делегат
Console.WriteLine("\nМультикаст делегат:");
AnimalDisplayDelegate multiDelegate = dog.DisplayInfo;
multiDelegate += cat.DisplayInfo;
multiDelegate += bird.DisplayInfo;
multiDelegate();

// Тестируем лямбда-выражения
Console.WriteLine("\n4. ЛЯМБДА-ВЫРАЖЕНИЯ:");
List<Animal> animals = new List<Animal> { dog, cat, bird };

// Фильтрация с лямбдой
Console.WriteLine("Животные младше 3 лет:");
var youngAnimals = animals.Where(a => a.Age < 3);
foreach (var animal in youngAnimals)
{
    animal.DisplayInfo();
}

// Сортировка с лямбдой
Console.WriteLine("\nЖивотные, отсортированные по имени:");
var sortedAnimals = animals.OrderBy(a => a.Name);
foreach (var animal in sortedAnimals)
{
    Console.WriteLine($"- {animal.Name}");
}

// Тестируем события
Console.WriteLine("\n5. СОБЫТИЯ:");

// Подписываемся на события
dog.AgeChanged += (message) => Console.WriteLine($"Событие: {message}");
cat.AgeChanged += OnAgeChanged;
bird.AgeChanged += (msg) => Console.WriteLine($"Уведомление о птице: {msg}");

// Изменяем возраст - вызовет события
Console.WriteLine("Изменяем возраст животных:");
dog.SetAge(4);
cat.SetAge(3);
bird.SetAge(2);

// Вспомогательный метод для события
void OnAgeChanged(string message)
{
    Console.WriteLine($"Обработчик события: {message}");
}

Console.WriteLine("\n=== ТЕСТ ЗАВЕРШЕН ===");

// Дополнительные тесты
Console.WriteLine("\n6. ДОПОЛНИТЕЛЬНЫЕ ПРИМЕРЫ:");

// Использование Action делегата
Action makeAllSounds = dog.MakeSound;
makeAllSounds += cat.MakeSound;
makeAllSounds += bird.MakeSound;

Console.WriteLine("Все звуки через Action:");
makeAllSounds();

// Использование Func делегата
Func<Animal, string> getInfo = (a) => $"{a.Name} ({a.Species}, {a.Age} лет)";
Console.WriteLine("\nИнформация через Func:");
foreach (var animal in animals)
{
    Console.WriteLine(getInfo(animal));
}
````
