<h1 style="color:DodgerBlue">Коллекции в .NET </h1>



Коллекции в .NET - это структуры данных, которые позволяют организовать и управлять наборами объектов. Они предоставляют эффективные способы хранения, доступа и манипулирования данными. Вот основные типы коллекций в .NET:

### 1. Статические коллекции

Статические коллекции являются неизменяемыми и могут содержать только один экземпляр заданного типа данных.

- `System.Collections.Generic.EmptyCollection<T>`
- `System.Collections.Generic.Singleton<T>`
- `System.Collections.ObjectModel.ReadOnlyCollection<T>`

### 2. Неупорядоченные коллекции

Неупорядоченные коллекции не гарантируют порядок элементов.

- `System.Collections.Generic.List<T>` - динамический массив
- `System.Collections.Generic.Dictionary<TKey,TValue>` - словарь (ассоциативный массив)
- `System.Collections.Generic.HashSet<T>` - множество уникальных элементов
- `System.Collections.Generic.SortedSet<T>` - отсортированный набор элементов

### 3. Сортированные коллекции

Сортированные коллекции поддерживают упорядоченный доступ и могут быть отсортированы по определенному критерию.

- `System.Collections.Generic.SortedDictionary<TKey,TValue>`
- `System.Collections.Generic.SortedSet<T>`
- `System.Collections.Generic.SortedList<TKey,TValue>`

### 4. Множества

Множества содержат только уникальные элементы.

- `System.Collections.Generic.HashSet<T>`
- `System.Collections.Generic.ISet<T>` - интерфейс для неупорядоченных множеств

### 5. Ключевые коллекции

Ключевые коллекции используются для хранения пар ключ-значение.

- `System.Collections.Generic.Dictionary<TKey,TValue>`
- `System.Collections.Generic.KeyedCollection<TKey,TElement>`
- `System.Collections.Generic.Lookup<TKey,TValue>`

### 6. Коллекции с ограниченным размером

Эти коллекции имеют фиксированный или максимальный размер.

- `System.Collections.Generic.Stack<T>` - стек
- `System.Collections.Generic.Queue<T>` - очередь
- `System.Collections.Generic.ConcurrentQueue<T>` - потокобезопасная очередь

### 7. Потокобезопасные коллекции

Потокобезопасные коллекции обеспечивают безопасность доступа из нескольких потоков.

- `System.Collections.Concurrent.ConcurrentDictionary<TKey,TValue>`
- `System.Collections.Concurrent.ConcurrentBag<T>`
- `System.Collections.Concurrent.BlockingCollection<T>`

### 8. Специализированные коллекции

Специализированные коллекции предназначены для конкретных задач.

- `System.Collections.Generic.BCLPortabilityLibrary.Collections.Generic.MultiMap<TKey,TKey2,TValue>`
- `System.Collections.Generic.BCLPortabilityLibrary.Collections.Generic.MultiDictionary<TKey,TValue>`

Это основные типы коллекций в .NET Framework и .NET Core. Выбор конкретной коллекции зависит от требований вашего проекта, таких как необходимость сортировки, потокобезопасности или специфики использования данных.

Работа с `List<T>` и `Dictionary<TKey, TValue>` в C# позволяет эффективно управлять коллекциями объектов различных типов. Давайте рассмотрим, как это можно сделать на примере классов `Transport`, `Car`, и `Bike`.

### Класс Transport и его производные

Начнем с базового класса `Transport`, который будет основой для классов `Car` и `Bike`.

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет наситель: {HasCarrier}");
    }
}
```

### Работа с List<T>

Теперь давайте посмотрим, как работать с `List<T>` для хранения коллекции объектов транспорта.

```csharp
using System;
using System.Collections.Generic;


// Создаем список Transport
List<Transport> transportList = new List<Transport>();

// Добавляем объекты Car и Bike в список
transportList.Add(new Car("Model S", "Tesla", 4));
transportList.Add(new Bike("Mountain Explorer", "Giant", true));

// Перебираем и выводим информацию обо всех транспортных средствах
foreach (var transport in transportList)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

```

### Работа с Dictionary<TKey, TValue>

Теперь давайте создадим коллекцию, которая будет хранить объекты `Transport`, используя `Dictionary<TKey, TValue>`, где ключом будет, например, регистрационный номер транспорта.

```csharp
using System;
using System.Collections.Generic;

// Создаем словарь с ключом типа string и значением типа Transport
Dictionary<string, Transport> transportDictionary = new Dictionary<string, Transport>();

// Добавляем объекты с уникальными ключами
transportDictionary.Add("TSL123", new Car("Model S", "Tesla", 4));
transportDictionary.Add("BIK456", new Bike("Mountain Explorer", "Giant", true));

// Поиск и вывод информации о транспорте по ключу
if (transportDictionary.TryGetValue("TSL123", out Transport foundTransport))
{
    Console.WriteLine("Найденый траснпорт:");
    foundTransport.DisplayInfo();
}
else
{
    Console.WriteLine("Транспорт не найден");
}

```

### Объяснения и примеры использования:

- **List<T>**: Используется, когда порядок элементов важен и возможны дубликаты. Вы можете добавлять, удалять, итерировать и изменять элементы по индексу.

- **Dictionary<TKey,TValue>**: Подходит, когда нужно организовать элементы по уникальным ключам для быстрого доступа. Это удобно для случаев, когда вы хотите находить элемент по ключу, как регистрационный номер в примере.

Эти структуры данных предоставляют мощные и гибкие возможности для управления коллекциями объектов, каждая из которых имеет свои преимущества в определённых ситуациях.

````
using System;
using System.Collections.Generic;

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет наситель: {HasCarrier}");
    }
}

Console.WriteLine("=== ПРОВЕРКА РАБОТЫ С КОЛЛЕКЦИЯМИ ===\n");

Console.WriteLine("1. РАБОТА С LIST<T>:\n");
List<Transport> transportList = new List<Transport>();

transportList.Add(new Car("Model S", "Tesla", 4));
transportList.Add(new Bike("Mountain Explorer", "Giant", true));
transportList.Add(new Car("Camry", "Toyota", 4));
transportList.Add(new Bike("Road Master", "Trek", false));

Console.WriteLine("Все транспортные средства:");
foreach (var transport in transportList)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

Console.WriteLine("Автомобили с 4 дверями:");
var carsWith4Doors = transportList.FindAll(t => t is Car car && car.NumberOfDoors == 4);
foreach (var car in carsWith4Doors)
{
    car.DisplayInfo();
}

Console.WriteLine("\n2. РАБОТА С DICTIONARY<TKey, TValue>:\n");
Dictionary<string, Transport> transportDictionary = new Dictionary<string, Transport>();

transportDictionary.Add("TSL123", new Car("Model S", "Tesla", 4));
transportDictionary.Add("BIK456", new Bike("Mountain Explorer", "Giant", true));
transportDictionary.Add("TOY789", new Car("Camry", "Toyota", 4));
transportDictionary.Add("TRE012", new Bike("Road Master", "Trek", false));

Console.WriteLine("Поиск транспорта по ключу:");
if (transportDictionary.TryGetValue("TSL123", out Transport foundTransport))
{
    Console.WriteLine("Найденный транспорт:");
    foundTransport.DisplayInfo();
}
else
{
    Console.WriteLine("Транспорт не найден");
}


Console.WriteLine("\nВсе транспортные средства в словаре:");
foreach (var kvp in transportDictionary)
{
    Console.WriteLine($"Ключ: {kvp.Key}");
    kvp.Value.DisplayInfo();
    Console.WriteLine();
}

Console.WriteLine("3. КОМБИНАЦИЯ LIST И DICTIONARY:\n");

Dictionary<Type, List<Transport>> groupedTransport = new Dictionary<Type, List<Transport>>();

foreach (var transport in transportList)
{
    Type transportType = transport.GetType();
    if (!groupedTransport.ContainsKey(transportType))
    {
        groupedTransport[transportType] = new List<Transport>();
    }
    groupedTransport[transportType].Add(transport);
}

Console.WriteLine("Транспорт сгруппированный по типу:");
foreach (var group in groupedTransport)
{
    Console.WriteLine($"\nТип: {group.Key.Name}");
    foreach (var transport in group.Value)
    {
        transport.DisplayInfo();
    }
}
````

````
using System;
using System.Collections.Generic;
using System.Linq;

// Базовый класс Animal
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }

    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Животное: {Name}, Возраст: {Age} лет, Вес: {Weight} кг");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, double weight, string breed)
        : base(name, age, weight)
    {
        Breed = breed;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }

    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит палку!");
    }
}

// Производный класс Cat
public class Cat : Animal
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }

    public Cat(string name, int age, double weight, string color, bool isIndoor)
        : base(name, age, weight)
    {
        Color = color;
        IsIndoor = isIndoor;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Цвет: {Color}, Домашний: {IsIndoor}");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }

    public Bird(string name, int age, double weight, double wingspan, bool canFly)
        : base(name, age, weight)
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan} см, Умеет летать: {CanFly}");
    }

    public void Fly()
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} летает!");
        }
        else
        {
            Console.WriteLine($"{Name} не умеет летать");
        }
    }
}

// Производный класс Fish
public class Fish : Animal
{
    public string WaterType { get; set; }
    public double SwimSpeed { get; set; }

    public Fish(string name, int age, double weight, string waterType, double swimSpeed)
        : base(name, age, weight)
    {
        WaterType = waterType;
        SwimSpeed = swimSpeed;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} издает булькающие звуки");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}, Скорость плавания: {SwimSpeed} км/ч");
    }

    public void Swim()
    {
        Console.WriteLine($"{Name} плывет со скоростью {SwimSpeed} км/ч");
    }
}

// 
Console.WriteLine("=== РАБОТА С КОЛЛЕКЦИЯМИ ANIMAL ===\n");

// 1. РАБОТА С LIST<T>
Console.WriteLine("1. LIST<ANIMAL> - ДИНАМИЧЕСКИЙ СПИСОК ЖИВОТНЫХ:\n");
List<Animal> animalList = new List<Animal>();

animalList.Add(new Dog("Шарик", 3, 12.5, "Овчарка"));
animalList.Add(new Cat("Барсик", 2, 4.2, "рыжий", true));
animalList.Add(new Bird("Кеша", 1, 0.3, 15.5, true));
animalList.Add(new Fish("Немо", 1, 0.5, "морская", 2.5));
animalList.Add(new Dog("Рекс", 4, 15.0, "Доберман"));
animalList.Add(new Cat("Мурка", 3, 3.8, "серая", false));

Console.WriteLine("Все животные в списке:");
foreach (var animal in animalList)
{
    animal.DisplayInfo();
    animal.MakeSound();
    Console.WriteLine();
}

Console.WriteLine("Собаки тяжелее 10 кг:");
var heavyDogs = animalList.Where(a => a is Dog dog && dog.Weight > 10).ToList();
foreach (var dog in heavyDogs)
{
    dog.DisplayInfo();
}

Console.WriteLine("\nЖивотные отсортированные по имени:");
var sortedAnimals = animalList.OrderBy(a => a.Name).ToList();
foreach (var animal in sortedAnimals)
{
    Console.WriteLine($"- {animal.Name}");
}

// 2. РАБОТА С DICTIONARY
Console.WriteLine("\n2. DICTIONARY<string, Animal> - СПРАВОЧНИК ЖИВОТНЫХ ПО ID:\n");
Dictionary<string, Animal> animalDictionary = new Dictionary<string, Animal>();

animalDictionary.Add("DOG001", new Dog("Шарик", 3, 12.5, "Овчарка"));
animalDictionary.Add("CAT001", new Cat("Барсик", 2, 4.2, "рыжий", true));
animalDictionary.Add("BRD001", new Bird("Кеша", 1, 0.3, 15.5, true));
animalDictionary.Add("FSH001", new Fish("Немо", 1, 0.5, "морская", 2.5));
animalDictionary.Add("DOG002", new Dog("Рекс", 4, 15.0, "Доберман"));
animalDictionary.Add("CAT002", new Cat("Мурка", 3, 3.8, "серая", false));

Console.WriteLine("Поиск животных по ID:");
if (animalDictionary.TryGetValue("DOG001", out Animal foundAnimal))
{
    Console.WriteLine("Найденное животное:");
    foundAnimal.DisplayInfo();
}
else
{
    Console.WriteLine("Животное не найдено");
}

Console.WriteLine("\nВсе животные в словаре:");
foreach (var kvp in animalDictionary)
{
    Console.WriteLine($"ID: {kvp.Key}");
    kvp.Value.DisplayInfo();
    Console.WriteLine();
}

// 3. ГРУППИРОВКА С ПОМОЩЬЮ DICTIONARY
Console.WriteLine("3. ГРУППИРОВКА ЖИВОТНЫХ ПО ТИПУ:\n");
Dictionary<Type, List<Animal>> groupedAnimals = new Dictionary<Type, List<Animal>>();

foreach (var animal in animalList)
{
    Type animalType = animal.GetType();
    if (!groupedAnimals.ContainsKey(animalType))
    {
        groupedAnimals[animalType] = new List<Animal>();
    }
    groupedAnimals[animalType].Add(animal);
}

Console.WriteLine("Животные сгруппированные по типу:");
foreach (var group in groupedAnimals)
{
    Console.WriteLine($"\n--- {group.Key.Name} ({group.Value.Count} шт.) ---");
    foreach (var animal in group.Value)
    {
        animal.DisplayInfo();
    }
}

// 4. СЛОЖНЫЕ ОПЕРАЦИИ С КОЛЛЕКЦИЯМИ
Console.WriteLine("\n4. СЛОЖНЫЕ ОПЕРАЦИИ С КОЛЛЕКЦИЯМИ:\n");

Dictionary<string, Animal> nameDictionary = animalList.ToDictionary(a => a.Name);

Console.WriteLine("Поиск животного по имени 'Барсик':");
if (nameDictionary.ContainsKey("Барсик"))
{
    nameDictionary["Барсик"].DisplayInfo();
}

// Объединение коллекций
Console.WriteLine("\nОбъединение животных из List и Dictionary:");
HashSet<Animal> allAnimals = new HashSet<Animal>(animalList);
allAnimals.UnionWith(animalDictionary.Values);

Console.WriteLine($"Всего уникальных животных: {allAnimals.Count}");

Console.WriteLine("\n5. ПРОДВИНУТЫЕ ВОЗМОЖНОСТИ КОЛЛЕКЦИЙ:\n");

Console.WriteLine("Очередь животных на ветеринарный осмотр:");
Queue<Animal> vetQueue = new Queue<Animal>();
vetQueue.Enqueue(new Dog("Бобик", 2, 8.0, "Дворняжка"));
vetQueue.Enqueue(new Cat("Васька", 1, 3.5, "черный", true));
vetQueue.Enqueue(new Bird("Соловей", 1, 0.2, 12.0, true));

while (vetQueue.Count > 0)
{
    var nextAnimal = vetQueue.Dequeue();
    Console.WriteLine($"Осмотр: {nextAnimal.Name}");
}

Console.WriteLine("\nСтек задач по уходу за животными:");
Stack<string> careTasks = new Stack<string>();
careTasks.Push("Покормить рыб");
careTasks.Push("Выгулять собак");
careTasks.Push("Почистить клетки птиц");
careTasks.Push("Поменять воду котам");

while (careTasks.Count > 0)
{
    Console.WriteLine($"Задача: {careTasks.Pop()}");
}

Console.WriteLine("\n6. LINQ ЗАПРОСЫ К КОЛЛЕКЦИЯМ ЖИВОТНЫХ:\n");

var ageGroups = from animal in animalList
                group animal by animal.Age into ageGroup
                orderby ageGroup.Key
                select new { Age = ageGroup.Key, Animals = ageGroup.ToList() };

Console.WriteLine("Животные сгруппированные по возрасту:");
foreach (var group in ageGroups)
{
    Console.WriteLine($"\nВозраст {group.Age} лет:");
    foreach (var animal in group.Animals)
    {
        Console.WriteLine($"- {animal.Name} ({animal.GetType().Name})");
    }
}


double averageWeight = animalList.Average(a => a.Weight);
var heavyAnimals = animalList.Where(a => a.Weight > averageWeight);

Console.WriteLine($"\nЖивотные тяжелее среднего веса ({averageWeight:F1} кг):");
foreach (var animal in heavyAnimals)
{
    Console.WriteLine($"- {animal.Name}: {animal.Weight} кг");
}

Console.WriteLine("\n7. СТАТИСТИКА ПО КОЛЛЕКЦИИ ЖИВОТНЫХ:\n");
Console.WriteLine($"Всего животных в List: {animalList.Count}");
Console.WriteLine($"Всего животных в Dictionary: {animalDictionary.Count}");
Console.WriteLine($"Количество собак: {animalList.Count(a => a is Dog)}");
Console.WriteLine($"Количество котов: {animalList.Count(a => a is Cat)}");
Console.WriteLine($"Количество птиц: {animalList.Count(a => a is Bird)}");
Console.WriteLine($"Количество рыб: {animalList.Count(a => a is Fish)}");
Console.WriteLine($"Средний вес: {animalList.Average(a => a.Weight):F1} кг");
Console.WriteLine($"Самый старый: {animalList.OrderByDescending(a => a.Age).First().Name} ({animalList.Max(a => a.Age)} лет)");
Console.WriteLine($"Самый молодой: {animalList.OrderBy(a => a.Age).First().Name} ({animalList.Min(a => a.Age)} лет)");
````
