```
// Базовый класс Book
public class Book
{
    // Приватные поля для инкапсуляции
    private string _title;
    private string _author;
    private int _yearOfPublication;
    private bool _isBorrowed;
    private Reader _currentReader;

    // Конструктор с использованием свойств
    public Book(string title, string author, int yearOfPublication)
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
        IsBorrowed = false;
        CurrentReader = null;
    }

    // Свойство Title с геттером и сеттером
    public string Title
    {
        get { return _title; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _title = value;
            else
                throw new ArgumentException("Название книги не может быть пустым!");
        }
    }

    // Свойство Author с геттером и сеттером
    public string Author
    {
        get { return _author; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _author = value;
            else
                throw new ArgumentException("Автор не может быть пустым!");
        }
    }

    // Свойство YearOfPublication с геттером и сеттером
    public int YearOfPublication
    {
        get { return _yearOfPublication; }
        set 
        { 
            if (value > 0 && value <= DateTime.Now.Year)
                _yearOfPublication = value;
            else
                throw new ArgumentOutOfRangeException("Год издания должен быть положительным и не больше текущего года!");
        }
    }

    // Свойство IsBorrowed с геттером и сеттером
    public bool IsBorrowed
    {
        get { return _isBorrowed; }
        private set { _isBorrowed = value; }
    }

    // Свойство CurrentReader с геттером и сеттером
    public Reader CurrentReader
    {
        get { return _currentReader; }
        private set { _currentReader = value; }
    }

    // Методы
    public virtual string GetInfo()
    {
        string status = IsBorrowed ? $" (Взята читателем: {CurrentReader.Name})" : " (Доступна)";
        return $"Книга: '{Title}', Автор: {Author}, Год издания: {YearOfPublication}{status}";
    }

    public virtual void Read()
    {
        Console.WriteLine($"Чтение книги: '{Title}'");
    }

    public virtual void Borrow(Reader reader)
    {
        if (IsBorrowed)
        {
            Console.WriteLine($"Книга '{Title}' уже взята читателем {CurrentReader.Name}");
            return;
        }

        IsBorrowed = true;
        CurrentReader = reader;
        Console.WriteLine($"Книга '{Title}' выдана читателю {reader.Name}");
        
        // Взаимодействие с объектом Reader
        reader.AddBorrowedBook(this);
    }

    public void Return()
    {
        if (!IsBorrowed)
        {
            Console.WriteLine($"Книга '{Title}' уже возвращена");
            return;
        }

        Console.WriteLine($"Книга '{Title}' возвращена читателем {CurrentReader.Name}");
        IsBorrowed = false;
        
        // Взаимодействие с объектом Reader
        CurrentReader.RemoveBorrowedBook(this);
        CurrentReader = null;
    }

    // Метод для проверки доступности книги
    public bool IsAvailable()
    {
        return !IsBorrowed;
    }
}

// Производный класс Textbook (Учебник)
public class Textbook : Book
{
    private string _subject;

    // Конструктор
    public Textbook(string title, string author, int yearOfPublication, string subject) 
        : base(title, author, yearOfPublication)
    {
        Subject = subject;
    }

    // Свойство Subject с геттером и сеттером
    public string Subject
    {
        get { return _subject; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _subject = value;
            else
                throw new ArgumentException("Предмет не может быть пустым!");
        }
    }

    // Переопределение метода Read()
    public override void Read()
    {
        Console.WriteLine($"Изучение учебника по предмету '{Subject}': '{Title}'");
    }

    // Переопределение метода GetInfo()
    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}";
    }

    // Переопределение метода Borrow()
    public override void Borrow(Reader reader)
    {
        base.Borrow(reader);
        Console.WriteLine($"Учебник по предмету '{Subject}' требует внимательного изучения");
    }
}

// Производный класс Fiction (Художественная литература)
public class Fiction : Book
{
    private string _genre;

    // Конструктор
    public Fiction(string title, string author, int yearOfPublication, string genre) 
        : base(title, author, yearOfPublication)
    {
        Genre = genre;
    }

    // Свойство Genre с геттером и сеттером
    public string Genre
    {
        get { return _genre; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _genre = value;
            else
                throw new ArgumentException("Жанр не может быть пустым!");
        }
    }

    // Переопределение метода Borrow()
    public override void Borrow(Reader reader)
    {
        base.Borrow(reader);
        Console.WriteLine($"Художественная книга в жанре '{Genre}' выдана для удовольствия чтения");
    }

    // Переопределение метода GetInfo()
    public override string GetInfo()
    {
        return base.GetInfo() + $", Жанр: {Genre}";
    }

    // Специфичный метод для художественной литературы
    public void RecommendSimilarBooks(Fiction[] books)
    {
        Console.WriteLine($"Похожие книги в жанре '{Genre}':");
        foreach (var book in books)
        {
            if (book.Genre == Genre && book.Title != Title)
            {
                Console.WriteLine($"- {book.Title} ({book.Author})");
            }
        }
    }
}

// Производный класс ScientificLiterature (Научная литература)
public class ScientificLiterature : Book
{
    private string _fieldOfScience;

    // Конструктор
    public ScientificLiterature(string title, string author, int yearOfPublication, string fieldOfScience) 
        : base(title, author, yearOfPublication)
    {
        FieldOfScience = fieldOfScience;
    }

    // Свойство FieldOfScience с геттером и сеттером
    public string FieldOfScience
    {
        get { return _fieldOfScience; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _fieldOfScience = value;
            else
                throw new ArgumentException("Область науки не может быть пустой!");
        }
    }

    // Переопределение метода GetInfo()
    public override string GetInfo()
    {
        return base.GetInfo() + $", Область науки: {FieldOfScience}";
    }

    // Переопределение метода Read()
    public override void Read()
    {
        Console.WriteLine($"Исследование научной работы в области '{FieldOfScience}': '{Title}'");
    }

    // Переопределение метода Borrow()
    public override void Borrow(Reader reader)
    {
        base.Borrow(reader);
        Console.WriteLine($"Научная литература требует специальных знаний в области '{FieldOfScience}'");
    }
}

// Класс для читателя (взаимодействие между объектами)
public class Reader
{
    private string _name;
    private string _email;
    private Book[] _borrowedBooks;
    private int _bookCount;

    // Конструктор
    public Reader(string name, string email)
    {
        Name = name;
        Email = email;
        _borrowedBooks = new Book[10];
        _bookCount = 0;
    }

    // Свойство Name с геттером и сеттером
    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _name = value;
            else
                throw new ArgumentException("Имя читателя не может быть пустым!");
        }
    }

    // Свойство Email с геттером и сеттером
    public string Email
    {
        get { return _email; }
        set 
        { 
            if (!string.IsNullOrEmpty(value) && value.Contains("@"))
                _email = value;
            else
                throw new ArgumentException("Email должен быть корректным!");
        }
    }

    // Метод для добавления взятой книги
    public void AddBorrowedBook(Book book)
    {
        if (_bookCount < _borrowedBooks.Length)
        {
            _borrowedBooks[_bookCount] = book;
            _bookCount++;
        }
        else
        {
            Console.WriteLine($"{Name} уже взял максимальное количество книг");
        }
    }

    // Метод для удаления возвращенной книги
    public void RemoveBorrowedBook(Book book)
    {
        for (int i = 0; i < _bookCount; i++)
        {
            if (_borrowedBooks[i] == book)
            {
                for (int j = i; j < _bookCount - 1; j++)
                {
                    _borrowedBooks[j] = _borrowedBooks[j + 1];
                }
                _bookCount--;
                break;
            }
        }
    }

    // Метод для отображения взятых книг
    public void DisplayBorrowedBooks()
    {
        Console.WriteLine($"\nКниги, взятые {Name}:");
        if (_bookCount == 0)
        {
            Console.WriteLine("Нет взятых книг");
            return;
        }

        for (int i = 0; i < _bookCount; i++)
        {
            Console.WriteLine($"- {_borrowedBooks[i].Title}");
        }
    }

    // Метод для поиска книги по названию
    public Book FindBorrowedBook(string title)
    {
        for (int i = 0; i < _bookCount; i++)
        {
            if (_borrowedBooks[i].Title == title)
            {
                return _borrowedBooks[i];
            }
        }
        return null;
    }
}

// Класс библиотеки для управления взаимодействием
public class Library
{
    private Book[] _books;
    private Reader[] _readers;
    private int _bookCount;
    private int _readerCount;

    // Конструктор
    public Library()
    {
        _books = new Book[100];
        _readers = new Reader[50];
        _bookCount = 0;
        _readerCount = 0;
    }

    // Метод для добавления книги в библиотеку
    public void AddBook(Book book)
    {
        if (_bookCount < _books.Length)
        {
            _books[_bookCount] = book;
            _bookCount++;
        }
    }

    // Метод для регистрации читателя
    public void RegisterReader(Reader reader)
    {
        if (_readerCount < _readers.Length)
        {
            _readers[_readerCount] = reader;
            _readerCount++;
        }
    }

    // Метод для выдачи книги
    public void BorrowBook(string bookTitle, string readerName)
    {
        Book book = FindBook(bookTitle);
        Reader reader = FindReader(readerName);

        if (book == null || reader == null)
        {
            Console.WriteLine("Книга или читатель не найдены");
            return;
        }

        if (book.IsAvailable())
        {
            book.Borrow(reader);
        }
        else
        {
            Console.WriteLine($"Книга '{bookTitle}' уже взята другим читателем");
        }
    }

    // Метод для возврата книги
    public void ReturnBook(string bookTitle, string readerName)
    {
        Book book = FindBook(bookTitle);
        Reader reader = FindReader(readerName);

        if (book == null || reader == null)
        {
            Console.WriteLine("Книга или читатель не найдены");
            return;
        }

        if (!book.IsAvailable() && book.IsBorrowed)
        {
            book.Return();
        }
        else
        {
            Console.WriteLine($"Книга '{bookTitle}' не была взята читателем {readerName}");
        }
    }

    // Поиск книги по названию
    private Book FindBook(string title)
    {
        for (int i = 0; i < _bookCount; i++)
        {
            if (_books[i].Title == title)
            {
                return _books[i];
            }
        }
        return null;
    }

    // Поиск читателя по имени
    private Reader FindReader(string name)
    {
        for (int i = 0; i < _readerCount; i++)
        {
            if (_readers[i].Name == name)
            {
                return _readers[i];
            }
        }
        return null;
    }

    // Метод для отображения всех книг
    public void DisplayAllBooks()
    {
        Console.WriteLine("\n=== ВСЕ КНИГИ В БИБЛИОТЕКЕ ===");
        for (int i = 0; i < _bookCount; i++)
        {
            Console.WriteLine(_books[i].GetInfo());
        }
    }

    // Метод для отображения всех читателей
    public void DisplayAllReaders()
    {
        Console.WriteLine("\n=== ЗАРЕГИСТРИРОВАННЫЕ ЧИТАТЕЛИ ===");
        for (int i = 0; i < _readerCount; i++)
        {
            Console.WriteLine($"Читатель: {_readers[i].Name}, Email: {_readers[i].Email}");
            _readers[i].DisplayBorrowedBooks();
        }
    }
}

// Демонстрация работы с взаимодействием объектов
Console.WriteLine("=== ДЕМОНСТРАЦИЯ БИБЛИОТЕЧНОЙ СИСТЕМЫ ===\n");

// Создание библиотеки
Library library = new Library();

// Создание книг с использованием конструкторов
Book book1 = new Book("Основы программирования", "Иван Петров", 2020);
Textbook mathTextbook = new Textbook("Математика для начинающих", "Анна Сидорова", 2019, "Математика");
Fiction novel1 = new Fiction("Преступление и наказание", "Федор Достоевский", 1866, "Роман");
Fiction novel2 = new Fiction("1984", "Джордж Оруэлл", 1949, "Антиутопия");
ScientificLiterature physicsBook = new ScientificLiterature("Теория относительности", "Альберт Эйнштейн", 1916, "Физика");

// Добавление книг в библиотеку
library.AddBook(book1);
library.AddBook(mathTextbook);
library.AddBook(novel1);
library.AddBook(novel2);
library.AddBook(physicsBook);

// Создание читателей
Reader reader1 = new Reader("Алексей", "alex@mail.com");
Reader reader2 = new Reader("Мария", "maria@mail.com");

// Регистрация читателей в библиотеке
library.RegisterReader(reader1);
library.RegisterReader(reader2);

// Отображение начального состояния
library.DisplayAllBooks();
library.DisplayAllReaders();

Console.WriteLine("\n=== ВЗАИМОДЕЙСТВИЕ: ВЫДАЧА КНИГ ===");
// Взаимодействие между объектами: читатели берут книги
library.BorrowBook("Математика для начинающих", "Алексей");
library.BorrowBook("Преступление и наказание", "Мария");
library.BorrowBook("1984", "Алексей"); // Попытка взять вторую книгу

Console.WriteLine("\n=== СОСТОЯНИЕ ПОСЛЕ ВЫДАЧИ ===");
library.DisplayAllBooks();
library.DisplayAllReaders();

Console.WriteLine("\n=== ВЗАИМОДЕЙСТВИЕ: ВОЗВРАТ КНИГ ===");
// Взаимодействие между объектами: возврат книг
library.ReturnBook("Математика для начинающих", "Алексей");
library.ReturnBook("Преступление и наказание", "Мария");

Console.WriteLine("\n=== СОСТОЯНИЕ ПОСЛЕ ВОЗВРАТА ===");
library.DisplayAllBooks();
library.DisplayAllReaders();

Console.WriteLine("\n=== СПЕЦИФИЧНОЕ ВЗАИМОДЕЙСТВИЕ ===");
// Демонстрация специфического взаимодействия
Fiction[] allFiction = new Fiction[] { novel1, novel2 };
novel1.RecommendSimilarBooks(allFiction);

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ГЕТТЕРОВ И СЕТТЕРОВ ===");
// Демонстрация работы свойств с валидацией
try
{
    Book invalidBook = new Book("", "Автор", 2020); // Вызовет исключение
}
catch (ArgumentException ex)
{
    Console.WriteLine($"Ошибка: {ex.Message}");
}

try
{
    Reader invalidReader = new Reader("Петр", "invalid-email"); // Вызовет исключение
}
catch (ArgumentException ex)
{
    Console.WriteLine($"Ошибка: {ex.Message}");
}

// Корректное использование свойств
novel1.Genre = "Классическая литература";
physicsBook.FieldOfScience = "Теоретическая физика";

Console.WriteLine("\nПосле изменений:");
Console.WriteLine(novel1.GetInfo());
Console.WriteLine(physicsBook.GetInfo());
```
