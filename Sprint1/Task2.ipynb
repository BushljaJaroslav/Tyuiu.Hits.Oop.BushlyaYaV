### Создание экземпляров классов (объектов)

В C#, экземпляр класса создается с помощью оператора `new`, который выделяет память для нового объекта и вызывает конструктор класса для инициализации этого объекта. Конструктор — это специальный метод в классе, который имеет то же имя, что и класс, и используется для установки начального состояния объекта.

#### Пример создания экземпляров классов Vehicle и Car

```csharp
Vehicle myVehicle = new Vehicle();
Car myCar = new Car();
```

В этом примере создаются два объекта: один для класса `Vehicle` и другой для класса `Car`. Конструкторы этих классов будут автоматически вызваны при создании объектов.

### Взаимодействие объектов между Собой

Взаимодействие объектов между собой означает, что один объект может вызывать методы другого объекта или обращаться к его свойствам (если они доступны). Это позволяет объектам обмениваться информацией и координировать свои действия.

#### Пример взаимодействия объектов

```csharp
myVehicle.Color = "Red";
myVehicle.Speed = 60;

myCar.Color = "Blue";
myCar.FuelLevel = 100;
myCar.Refuel(50); // Увеличиваем уровень топлива на 50
```

В этом примере мы устанавливаем свойства `Color` и `Speed` для объекта `myVehicle`, а также свойства `Color` и `FuelLevel` для объекта `myCar`. Затем мы вызываем метод `Refuel` у объекта `myCar`, чтобы увеличить его уровень топлива.

### Закрепление правил использования модификаторов доступа

Модификаторы доступа определяют уровень доступности членов класса из других частей кода. Важно правильно использовать модификаторы доступа для обеспечения безопасности и инкапсуляции данных.

#### Примеры модификаторов доступа в классах Vehicle и Car

```csharp
public class Vehicle
{
    public string Color { get; set; } // public свойство
    private int speed; // private поле

    protected void SetSpeed(int value) // protected метод
    {
        speed = value;
    }
}

public class Car : Vehicle
{
    private int fuelLevel; // private поле

    public void Refuel(int amount) // public метод
    {
        fuelLevel += amount;
    }

    protected void CheckSpeedLimit() // protected метод
    {
        if (speed > 120)
        {
            Console.WriteLine("Превышен скоростной режим!");
        }
    }
}
```

В этом примере:
- Свойство `Color` в классе `Vehicle` объявлено как `public`, что позволяет доступ к нему из любого места в коде.
- Поле `speed` в классе `Vehicle` объявлено как `private`, что ограничивает доступ к нему только внутри класса `Vehicle`.
- Метод `SetSpeed` в классе `Vehicle` объявлен как `protected`, что позволяет доступ к нему из класса `Vehicle` и его производных классов.
- Поле `fuelLevel` в классе `Car` объявлено как `private`, ограничивая доступ к нему только внутри класса `Car`.
- Методы `Refuel` и `CheckSpeedLimit` в классе `Car` демонстрируют использование модификаторов доступа для контроля над тем, как объекты могут взаимодействовать друг с другом.

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (2-3 например Dog, Cat и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также статические и обычными атрибуты и метода класса с реализацией взаимодействия между объектами.

````
using System;

// Базовый класс Animal
public class Animal
{
    // Приватные поля для инкапсуляции
    private string _name;
    private int _age;
    private double _weight;
    
    // Статический атрибут для отслеживания всех животных
    public static int TotalAnimals { get; private set; } = 0;
    
    // Статическое поле для хранения всех созданных животных
    private static Animal[] _allAnimals = new Animal[100];
    private static int _animalCount = 0;

    // Публичные свойства с геттерами и сеттерами
    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым!");
        }
    }

    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
        }
    }

    public double Weight
    {
        get { return _weight; }
        set 
        { 
            if (value > 0)
                _weight = Math.Round(value, 2);
            else
                throw new ArgumentOutOfRangeException("Вес должен быть положительным!");
        }
    }

    // Конструктор по умолчанию
    public Animal()
    {
        _name = "Безымянный";
        _age = 0;
        _weight = 1.0;
        RegisterAnimal(this);
    }

    // Параметризированный конструктор
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
        RegisterAnimal(this);
    }

    // Приватный статический метод для регистрации животных
    private static void RegisterAnimal(Animal animal)
    {
        TotalAnimals++;
        if (_animalCount < _allAnimals.Length)
        {
            _allAnimals[_animalCount] = animal;
            _animalCount++;
        }
    }

    // Статический метод для получения всех животных
    public static Animal[] GetAllAnimals()
    {
        Animal[] result = new Animal[_animalCount];
        Array.Copy(_allAnimals, result, _animalCount);
        return result;
    }

    // Статический метод для поиска животного по имени
    public static Animal FindAnimalByName(string name)
    {
        foreach (var animal in _allAnimals)
        {
            if (animal != null && animal.Name == name)
                return animal;
        }
        return null;
    }

    // Публичные методы для взаимодействия
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук");
    }

    public void Eat(string food)
    {
        Console.WriteLine($"{Name} ест {food}");
        Weight += 0.1;
    }

    public void Sleep()
    {
        Console.WriteLine($"{Name} спит");
    }

    // Метод для взаимодействия с другим животным
    public virtual void InteractWith(Animal otherAnimal)
    {
        Console.WriteLine($"{Name} взаимодействует с {otherAnimal.Name}");
    }

    // Защищенный метод для внутреннего использования
    protected void IncreaseAge()
    {
        _age++;
    }

    // Виртуальный метод для отображения информации
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Животное: {Name}, Возраст: {Age} лет, Вес: {Weight} кг");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    private string _breed;
    public static int TotalDogs { get; private set; } = 0;

    public string Breed
    {
        get { return _breed; }
        set { _breed = value; }
    }

    public Dog(string name, int age, double weight, string breed) : base(name, age, weight)
    {
        Breed = breed;
        TotalDogs++;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }

    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит палку!");
    }

    // Взаимодействие с другим животным (полиморфизм)
    public override void InteractWith(Animal otherAnimal)
    {
        if (otherAnimal is Dog)
        {
            Console.WriteLine($"{Name} играет с другой собакой {otherAnimal.Name}");
        }
        else if (otherAnimal is Cat)
        {
            Console.WriteLine($"{Name} гоняется за котом {otherAnimal.Name}");
        }
        else
        {
            base.InteractWith(otherAnimal);
        }
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
}

// Производный класс Cat
public class Cat : Animal
{
    private int _lives;
    public static int TotalCats { get; private set; } = 0;

    public int Lives
    {
        get { return _lives; }
        set 
        { 
            if (value >= 0 && value <= 9)
                _lives = value;
            else
                throw new ArgumentOutOfRangeException("У кота может быть от 0 до 9 жизней!");
        }
    }

    public Cat(string name, int age, double weight, int lives = 9) : base(name, age, weight)
    {
        Lives = lives;
        TotalCats++;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }

    // Взаимодействие с другим животным
    public override void InteractWith(Animal otherAnimal)
    {
        if (otherAnimal is Dog)
        {
            Console.WriteLine($"{Name} убегает от собаки {otherAnimal.Name}");
        }
        else if (otherAnimal is Cat)
        {
            Console.WriteLine($"{Name} дружелюбно трется о другого кота {otherAnimal.Name}");
        }
        else
        {
            base.InteractWith(otherAnimal);
        }
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Осталось жизней: {Lives}");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    private double _wingspan;
    public static int TotalBirds { get; private set; } = 0;

    public double Wingspan
    {
        get { return _wingspan; }
        set 
        { 
            if (value > 0)
                _wingspan = Math.Round(value, 2);
            else
                throw new ArgumentOutOfRangeException("Размах крыльев должен быть положительным!");
        }
    }

    public Bird(string name, int age, double weight, double wingspan) : base(name, age, weight)
    {
        Wingspan = wingspan;
        TotalBirds++;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    public void Fly()
    {
        Console.WriteLine($"{Name} летит с размахом крыльев {Wingspan} см!");
    }

    public override void InteractWith(Animal otherAnimal)
    {
        if (otherAnimal is Dog || otherAnimal is Cat)
        {
            Console.WriteLine($"{Name} улетает от {otherAnimal.Name}");
        }
        else
        {
            base.InteractWith(otherAnimal);
        }
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan} см");
    }
}

// Класс для управления взаимодействием животных
public class ZooKeeper
{
    public string Name { get; set; }

    public ZooKeeper(string name)
    {
        Name = name;
    }

    // Метод для организации взаимодействия между животными
    public void MakeAnimalsInteract(Animal animal1, Animal animal2)
    {
        Console.WriteLine($"\nСмотритель {Name} организует встречу:");
        Console.WriteLine($"--- {animal1.Name} и {animal2.Name} ---");
        
        animal1.InteractWith(animal2);
        animal2.InteractWith(animal1);
        
        Console.WriteLine("--- Встреча завершена ---");
    }

    // Метод для кормления животных
    public void FeedAnimal(Animal animal, string food)
    {
        Console.WriteLine($"Смотритель {Name} кормит {animal.Name} {food}");
        animal.Eat(food);
    }
}

// Демонстрация работы с объектами
Console.WriteLine("=== СОЗДАНИЕ ЭКЗЕМПЛЯРОВ КЛАССОВ ===\n");

// Создание объектов с помощью оператора new
Animal genericAnimal = new Animal("Неизвестное животное", 2, 5.0);
Dog dog1 = new Dog("Шарик", 3, 12.5, "Овчарка");
Dog dog2 = new Dog("Рекс", 4, 15.0, "Доберман");
Cat cat1 = new Cat("Барсик", 2, 4.2, 7);
Cat cat2 = new Cat("Мурка", 1, 3.8, 9);
Bird bird1 = new Bird("Кеша", 1, 0.3, 15.5);
Bird bird2 = new Bird("Гоша", 2, 0.4, 18.0);

Console.WriteLine("=== ИНФОРМАЦИЯ О СОЗДАННЫХ ОБЪЕКТАХ ===");
dog1.DisplayInfo();
cat1.DisplayInfo();
bird1.DisplayInfo();

Console.WriteLine("\n=== ВЗАИМОДЕЙСТВИЕ МЕЖДУ ОБЪЕКТАМИ ===");

// Создаем смотрителя зоопарка
ZooKeeper keeper = new ZooKeeper("Иван");

// Организуем взаимодействие между животными
keeper.MakeAnimalsInteract(dog1, cat1);
keeper.MakeAnimalsInteract(dog1, dog2);
keeper.MakeAnimalsInteract(cat1, cat2);
keeper.MakeAnimalsInteract(bird1, dog1);

Console.WriteLine("\n=== КОРМЛЕНИЕ ЖИВОТНЫХ ===");
keeper.FeedAnimal(dog1, "мясо");
keeper.FeedAnimal(cat1, "рыбу");
keeper.FeedAnimal(bird1, "зерно");

Console.WriteLine("\n=== ИЗМЕНЕНИЕ СОСТОЯНИЯ ОБЪЕКТОВ ===");
// Взаимодействие через свойства
dog1.Age = 4;
cat1.Weight = 4.5;
bird1.Name = "Чижик";

Console.WriteLine("После изменений:");
dog1.DisplayInfo();
cat1.DisplayInfo();
bird1.DisplayInfo();

Console.WriteLine("\n=== ПОИСК И ВЗАИМОДЕЙСТВИЕ ЧЕРЕЗ СТАТИЧЕСКИЕ МЕТОДЫ ===");
// Поиск животных по имени
Animal foundDog = Animal.FindAnimalByName("Шарик");
Animal foundCat = Animal.FindAnimalByName("Барсик");

if (foundDog != null && foundCat != null)
{
    Console.WriteLine("Найдены животные:");
    keeper.MakeAnimalsInteract(foundDog, foundCat);
}

Console.WriteLine("\n=== РАБОТА СО ВСЕМИ ЖИВОТНЫМИ ===");
Animal[] allAnimals = Animal.GetAllAnimals();
Console.WriteLine($"Всего животных в системе: {allAnimals.Length}");

foreach (var animal in allAnimals)
{
    if (animal != null)
    {
        Console.WriteLine($"- {animal.Name} ({animal.GetType().Name})");
    }
}

Console.WriteLine("\n=== СТАТИСТИКА ===");
Console.WriteLine($"Всего животных: {Animal.TotalAnimals}");
Console.WriteLine($"Собак: {Dog.TotalDogs}");
Console.WriteLine($"Котов: {Cat.TotalCats}");
Console.WriteLine($"Птиц: {Bird.TotalBirds}");

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ МОДИФИКАТОРОВ ДОСТУПА ===");
Console.WriteLine("public свойства: animal.Name = 'Новое имя' (доступно везде)");
Console.WriteLine("private поля: _name (доступно только внутри класса)");
Console.WriteLine("protected методы: IncreaseAge() (доступно в классе и наследниках)");
````
