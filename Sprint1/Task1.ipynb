### Структура и Объявление Класса в C#

В C#, класс объявляется с использованием ключевого слова `class`, за которым следует имя класса. Класс может содержать поля (данные), свойства (способ доступа к данным), методы (функциональность) и события. 

#### Свойства

Свойства в C# используются для инкапсуляции данных. Они предоставляют гибкий механизм для чтения, записи или вычисления значений приватных полей. Свойства могут иметь два акцессора: `get` (геттер) для чтения значения и `set` (сеттер) для его изменения.

#### Геттеры и Сеттеры

Геттеры и сеттеры являются частью свойств класса в C#. Геттер (`get`) используется для возврата значения свойства, а сеттер (`set`) — для установки нового значения. Сеттер может включать логику валидации или другие операции перед изменением значения поля.

### Пример Классов Vehicle и Car

Давайте рассмотрим примеры классов `Vehicle` и `Car`, чтобы проиллюстрировать использование свойств, геттеров и сеттеров, а также разницу между статическими и обычными атрибутами и методами.

#### Класс Vehicle

```csharp
public class Vehicle
{
    private string _color;
    private int _speed;

    // Свойство Color с геттером и сеттером
    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    // Свойство Speed с геттером и сеттером
    public int Speed
    {
        get { return _speed; }
        set
        {
            if (value >= 0)
                _speed = value;
            else
                throw new ArgumentOutOfRangeException("Скорость не может быть отрицательной!");
        }
    }

    // Статический атрибут
    public static int TotalVehicles { get; private set; } = 0;

    // Конструктор
    public Vehicle()
    {
        TotalVehicles++;
    }

    // Обычный метод
    public void Accelerate(int delta)
    {
        Speed += delta;
    }

    // Статический метод
    public static void DisplayTotalVehicles()
    {
        Console.WriteLine($"Общее количество транспортных средств: {TotalVehicles}");
    }
}
```

#### Класс Car

```csharp
public class Car : Vehicle
{
    private int _fuelLevel;

    // Свойство FuelLevel с геттером и сеттером
    public int FuelLevel
    {
        get { return _fuelLevel; }
        set { _fuelLevel = value; }
    }

    // Конструктор
    public Car() : base()
    {
        FuelLevel = 100; // Задаем начальный уровень топлива
    }

    // Обычный метод
    public void Refuel(int amount)
    {
        FuelLevel += amount;
    }
}
```

### Разница между Статическими и Обычными Атрибутами и Методами Класса

- **Статические атрибуты и методы** принадлежат самому классу, а не его экземплярам. Они могут быть вызваны без создания объекта класса. В примере выше `TotalVehicles` является статическим атрибутом, а `DisplayTotalVehicles()` — статическим методом класса `Vehicle`.
- **Обычные атрибуты и методы** принадлежат экземплярам класса и могут иметь разные значения для разных объектов. В примере выше `_color`, `_speed` и `_fuelLevel` являются обычными атрибутами, а методы `Accelerate()` и `Refuel()` — обычными методами.

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (2-3 например Dog, Cat и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также статические и обычными атрибуты и метода класса

````
using System;

// Базовый класс Animal
public class Animal
{
    // Приватные поля для инкапсуляции данных
    private string _name;
    private int _age;
    private double _weight;
    
    // Статический атрибут - принадлежит классу, а не экземплярам
    public static int TotalAnimals { get; private set; } = 0;
    
    // Статическое поле для хранения всех созданных животных
    public static string[] AllAnimalNames = new string[100];
    private static int _animalCount = 0;

    // Свойство Name с геттером и сеттером (инкапсуляция)
    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым!");
        }
    }

    // Свойство Age с валидацией в сеттере
    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
        }
    }

    // Свойство Weight с валидацией
    public double Weight
    {
        get { return _weight; }
        set 
        { 
            if (value > 0)
                _weight = value;
            else
                throw new ArgumentOutOfRangeException("Вес должен быть положительным!");
        }
    }

    // Конструктор по умолчанию
    public Animal()
    {
        _name = "Безымянный";
        _age = 0;
        _weight = 1.0;
        TotalAnimals++;
        RegisterAnimal(_name);
    }

    // Параметризированный конструктор
    public Animal(string name, int age, double weight)
    {
        Name = name;    // Используем свойство для валидации
        Age = age;      // Используем свойство для валидации
        Weight = weight; // Используем свойство для валидации
        TotalAnimals++;
        RegisterAnimal(name);
    }

    // Статический метод для регистрации животных
    private static void RegisterAnimal(string name)
    {
        if (_animalCount < AllAnimalNames.Length)
        {
            AllAnimalNames[_animalCount] = name;
            _animalCount++;
        }
    }

    // Статический метод - можно вызвать без создания объекта
    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"Общее количество животных: {TotalAnimals}");
    }

    // Статический метод для отображения всех имен животных
    public static void DisplayAllAnimalNames()
    {
        Console.WriteLine("Список всех животных:");
        for (int i = 0; i < _animalCount; i++)
        {
            if (!string.IsNullOrEmpty(AllAnimalNames[i]))
                Console.WriteLine($"- {AllAnimalNames[i]}");
        }
    }

    // Обычный метод - требует экземпляр класса
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук");
    }

    // Обычный метод
    public void Eat(string food)
    {
        Console.WriteLine($"{Name} ест {food}");
        _weight += 0.1; // Увеличиваем вес после еды
    }

    // Обычный метод
    public void Sleep()
    {
        Console.WriteLine($"{Name} спит");
    }

    // Виртуальный метод для переопределения в производных классах
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Животное: {Name}, Возраст: {Age} лет, Вес: {Weight} кг");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    private string _breed;
    
    // Статический атрибут только для собак
    public static int TotalDogs { get; private set; } = 0;

    // Свойство Breed
    public string Breed
    {
        get { return _breed; }
        set { _breed = value; }
    }

    // Конструктор
    public Dog(string name, int age, double weight, string breed) : base(name, age, weight)
    {
        Breed = breed;
        TotalDogs++;
    }

    // Переопределение метода
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }

    // Новый метод, специфичный для собак
    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит палку!");
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }

    // Статический метод только для собак
    public static void DisplayTotalDogs()
    {
        Console.WriteLine($"Общее количество собак: {TotalDogs}");
    }
}

// Производный класс Cat
public class Cat : Animal
{
    private int _lives;
    
    // Статический атрибут только для котов
    public static int TotalCats { get; private set; } = 0;

    // Свойство Lives с валидацией
    public int Lives
    {
        get { return _lives; }
        set 
        { 
            if (value >= 0 && value <= 9)
                _lives = value;
            else
                throw new ArgumentOutOfRangeException("У кота может быть от 0 до 9 жизней!");
        }
    }

    // Конструктор
    public Cat(string name, int age, double weight, int lives = 9) : base(name, age, weight)
    {
        Lives = lives;
        TotalCats++;
    }

    // Переопределение метода
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    // Новый метод, специфичный для котов
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Осталось жизней: {Lives}");
    }

    // Статический метод только для котов
    public static void DisplayTotalCats()
    {
        Console.WriteLine($"Общее количество котов: {TotalCats}");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    private double _wingspan;
    
    // Статический атрибут только для птиц
    public static int TotalBirds { get; private set; } = 0;

    // Свойство Wingspan
    public double Wingspan
    {
        get { return _wingspan; }
        set 
        { 
            if (value > 0)
                _wingspan = value;
            else
                throw new ArgumentOutOfRangeException("Размах крыльев должен быть положительным!");
        }
    }

    // Конструктор
    public Bird(string name, int age, double weight, double wingspan) : base(name, age, weight)
    {
        Wingspan = wingspan;
        TotalBirds++;
    }

    // Переопределение метода
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    // Новый метод, специфичный для птиц
    public void Fly()
    {
        Console.WriteLine($"{Name} летит с размахом крыльев {Wingspan} см!");
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan} см");
    }

    // Статический метод только для птиц
    public static void DisplayTotalBirds()
    {
        Console.WriteLine($"Общее количество птиц: {TotalBirds}");
    }
}

// Демонстрация работы классов
Console.WriteLine("=== ДЕМОНСТРАЦИЯ КЛАССОВ ЖИВОТНЫХ ===\n");

// Создание объектов разных классов
Animal genericAnimal = new Animal("Неизвестное животное", 2, 5.0);
Dog dog = new Dog("Шарик", 3, 12.5, "Овчарка");
Cat cat = new Cat("Барсик", 2, 4.2, 7);
Bird bird = new Bird("Кеша", 1, 0.3, 15.5);

// Демонстрация обычных методов
Console.WriteLine("=== ОБЫЧНЫЕ МЕТОДЫ ===");
genericAnimal.MakeSound();
dog.MakeSound();
cat.MakeSound();
bird.MakeSound();

Console.WriteLine("\n=== ИНФОРМАЦИЯ О ЖИВОТНЫХ ===");
genericAnimal.DisplayInfo();
dog.DisplayInfo();
cat.DisplayInfo();
bird.DisplayInfo();

Console.WriteLine("\n=== СПЕЦИФИЧНЫЕ МЕТОДЫ ===");
dog.Fetch();
cat.Purr();
bird.Fly();

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ СВОЙСТВ И ИНКАПСУЛЯЦИИ ===");
// Использование свойств с валидацией
try
{
    dog.Age = -5; // Вызовет исключение
}
catch (ArgumentOutOfRangeException ex)
{
    Console.WriteLine($"Ошибка: {ex.Message}");
}

// Корректное изменение свойств
dog.Age = 4;
cat.Weight = 4.5;
bird.Name = "Гоша";

Console.WriteLine("\nПосле изменений:");
dog.DisplayInfo();
cat.DisplayInfo();
bird.DisplayInfo();

Console.WriteLine("\n=== СТАТИЧЕСКИЕ МЕТОДЫ И АТРИБУТЫ ===");
// Вызов статических методов
Animal.DisplayTotalAnimals();
Dog.DisplayTotalDogs();
Cat.DisplayTotalCats();
Bird.DisplayTotalBirds();

Console.WriteLine("\n=== СПИСОК ВСЕХ ЖИВОТНЫХ ===");
Animal.DisplayAllAnimalNames();

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ РАЗНИЦЫ МЕЖДУ СТАТИЧЕСКИМИ И ОБЫЧНЫМИ ЧЛЕНАМИ ===");
Console.WriteLine("Статические методы вызываются через имя класса: Animal.DisplayTotalAnimals()");
Console.WriteLine("Обычные методы вызываются через экземпляр: dog.MakeSound()");
Console.WriteLine($"Статический атрибут общий для всех: Всего животных = {Animal.TotalAnimals}");
Console.WriteLine($"Обычные атрибуты уникальны для каждого объекта: Собака весит = {dog.Weight} кг");
````
