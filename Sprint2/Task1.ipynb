<h1 style="color:DodgerBlue">Сложное наследование</h1>

Сложное наследование в C# отличается от простого тем, что оно позволяет создавать более глубокие иерархии классов, где один класс может наследоваться от другого, который уже является производным классом. Это позволяет еще больше абстрагировать общие свойства и методы, делая код более модульным и легким для поддержки.

Давайте рассмотрим пример сложного наследования, используя классы `Vehicle`, `Car`, и `Motorcycle`, а также введем новый класс `SportsCar`, который будет наследоваться от `Car`.

### Шаг 1: Базовый класс Vehicle

Начнем с определения базового класса `Vehicle`, как и в предыдущем примере.

```csharp
public class Vehicle
{
    public string Make { get; set; }
    public string Model { get; set; }

    public Vehicle(string make, string model)
    {
        Make = make;
        Model = model;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Производитель: {Make}, Модель: {Model}");
    }
}
```

### Шаг 2: Производный класс Car

Теперь определим класс `Car`, который наследуется от `Vehicle`.

```csharp
public class Car : Vehicle
{
    public int Doors { get; set; }

    public Car(string make, string model, int doors) : base(make, model)
    {
        Doors = doors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Двери: {Doors}");
    }
}
```

### Шаг 3: Производный класс SportsCar

Введем класс `SportsCar`, который наследуется от `Car`, демонстрируя сложное наследование.

```csharp
public class SportsCar : Car
{
    public bool IsConvertible { get; set; }

    public SportsCar(string make, string model, int doors, bool isConvertible) 
        : base(make, model, doors)
    {
        IsConvertible = isConvertible;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Автомобиль с откидным верхом: {IsConvertible}");
    }
}
```

### Шаг 4: Производный класс Motorcycle

Определим класс `Motorcycle`, который также наследуется от `Vehicle`.

```csharp
public class Motorcycle : Vehicle
{
    public string EngineType { get; set; }

    public Motorcycle(string make, string model, string engineType) : base(make, model)
    {
        EngineType = engineType;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип двигателя: {EngineType}");
    }
}
```

### Шаг 5: Использование классов

Теперь мы можем создать экземпляры наших классов и использовать их.

```csharp
Car myCar = new Car("Toyota", "Corolla", 4);
myCar.DisplayInfo();

SportsCar mySportsCar = new SportsCar("Ferrari", "488 Spider", 2, true);
mySportsCar.DisplayInfo();

Motorcycle myMotorcycle = new Motorcycle("Ducati", "Panigale V4", "V4");
myMotorcycle.DisplayInfo();
```

### Объяснение кода

- **Базовый класс `Vehicle`** предоставляет основные свойства и методы для всех транспортных средств.
- **Класс `Car`** наследует от `Vehicle` и добавляет свойство `Doors`.
- **Класс `SportsCar`** наследует от `Car`, демонстрируя сложное наследование, и добавляет свойство `IsConvertible`.
- **Класс `Motorcycle`** также наследует от `Vehicle`, но имеет свое собственное свойство `EngineType`.

### Отличие сложного наследования от простого

- **Глубина иерархии**: Сложное наследование позволяет создавать более глубокие иерархии классов, где классы могут наследоваться друг от друга на несколько уровней вниз.
- **Переиспользование кода**: Сложное наследование увеличивает возможности для переиспользования кода, поскольку классы могут наследовать не только непосредственно от базового класса, но и от других производных классов.
- **Флексибельность**: Сложное наследование обеспечивает большую гибкость в проектировании программ, позволяя разработчикам точно определять отношения между классами и их роли в системе.

Таким образом, сложное наследование в C# представляет собой мощный инструмент для создания хорошо структурированных и масштабируемых приложений, позволяя эффективно организовывать код и избегать дублирования.

````
public class Vehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    public int Year { get; set; }

    public Vehicle(string make, string model, int year)
    {
        Make = make;
        Model = model;
        Year = year;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Транспорт: {Make} {Model}, Год: {Year}");
    }
}

// Производный класс Car
public class Car : Vehicle
{
    public int Doors { get; set; }
    public string FuelType { get; set; }

    public Car(string make, string model, int year, int doors, string fuelType) 
        : base(make, model, year)
    {
        Doors = doors;
        FuelType = fuelType;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Двери: {Doors}, Тип топлива: {FuelType}");
    }
}

// Производный класс SportsCar (сложное наследование)
public class SportsCar : Car
{
    public double TopSpeed { get; set; }
    public bool IsConvertible { get; set; }

    public SportsCar(string make, string model, int year, int doors, string fuelType, 
                    double topSpeed, bool isConvertible) 
        : base(make, model, year, doors, fuelType)
    {
        TopSpeed = topSpeed;
        IsConvertible = isConvertible;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Макс. скорость: {TopSpeed} км/ч, Кабриолет: {(IsConvertible ? "Да" : "Нет")}");
    }

    public void RaceMode()
    {
        Console.WriteLine($"{Make} {Model} включил гоночный режим!");
    }
}

// Производный класс Motorcycle
public class Motorcycle : Vehicle
{
    public string EngineType { get; set; }
    public bool HasSidecar { get; set; }

    public Motorcycle(string make, string model, int year, string engineType, bool hasSidecar) 
        : base(make, model, year)
    {
        EngineType = engineType;
        HasSidecar = hasSidecar;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип двигателя: {EngineType}, Коляска: {(HasSidecar ? "Есть" : "Нет")}");
    }
}

// Производный класс SportBike (сложное наследование от Motorcycle)
public class SportBike : Motorcycle
{
    public string RacingClass { get; set; }
    public bool HasABS { get; set; }

    public SportBike(string make, string model, int year, string engineType, bool hasSidecar,
                    string racingClass, bool hasABS) 
        : base(make, model, year, engineType, hasSidecar)
    {
        RacingClass = racingClass;
        HasABS = hasABS;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Гоночный класс: {RacingClass}, ABS: {(HasABS ? "Есть" : "Нет")}");
    }

    public void Wheelie()
    {
        Console.WriteLine($"{Make} {Model} делает вилли!");
    }
}

Console.WriteLine("=== Сложное наследование: Транспортные средства ===");

// Простое наследование
Car myCar = new Car("Toyota", "Corolla", 2022, 4, "Бензин");
myCar.DisplayInfo();

Console.WriteLine();

// Сложное наследование - SportsCar наследует от Car
SportsCar mySportsCar = new SportsCar("Ferrari", "488 Spider", 2023, 2, "Бензин", 330, true);
mySportsCar.DisplayInfo();
mySportsCar.RaceMode();

Console.WriteLine();

// Простое наследование
Motorcycle myMotorcycle = new Motorcycle("Harley-Davidson", "Street 750", 2021, "V-Twin", false);
myMotorcycle.DisplayInfo();

Console.WriteLine();

// Сложное наследование - SportBike наследует от Motorcycle
SportBike mySportBike = new SportBike("Ducati", "Panigale V4", 2023, "V4", false, "Superbike", true);
mySportBike.DisplayInfo();
mySportBike.Wheelie();
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также сложное наследование.

````
// Базовый класс Animal
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }

    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Животное: {Name}, Возраст: {Age}, Вес: {Weight}кг");
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук");
    }
}

// Производный класс Mammal
public class Mammal : Animal
{
    public string FurColor { get; set; }
    public bool IsDomestic { get; set; }

    public Mammal(string name, int age, double weight, string furColor, bool isDomestic) 
        : base(name, age, weight)
    {
        FurColor = furColor;
        IsDomestic = isDomestic;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Цвет шерсти: {FurColor}, Домашний: {(IsDomestic ? "Да" : "Нет")}");
    }

    public void FeedMilk()
    {
        Console.WriteLine($"{Name} кормит молоком своих детенышей");
    }
}

// Производный класс Dog (сложное наследование от Mammal)
public class Dog : Mammal
{
    public string Breed { get; set; }
    public string Size { get; set; }

    public Dog(string name, int age, double weight, string furColor, bool isDomestic,
              string breed, string size) 
        : base(name, age, weight, furColor, isDomestic)
    {
        Breed = breed;
        Size = size;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}, Размер: {Size}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }

    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит палку!");
    }
}

// Производный класс WorkingDog (сложное наследование от Dog)
public class WorkingDog : Dog
{
    public string Job { get; set; }
    public bool IsTrained { get; set; }

    public WorkingDog(string name, int age, double weight, string furColor, bool isDomestic,
                     string breed, string size, string job, bool isTrained) 
        : base(name, age, weight, furColor, isDomestic, breed, size)
    {
        Job = job;
        IsTrained = isTrained;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Работа: {Job}, Дрессирован: {(IsTrained ? "Да" : "Нет")}");
    }

    public void DoJob()
    {
        Console.WriteLine($"{Name} выполняет свою работу: {Job}");
    }
}

// Производный класс Cat (сложное наследование от Mammal)
public class Cat : Mammal
{
    public bool IsIndoor { get; set; }
    public string Personality { get; set; }

    public Cat(string name, int age, double weight, string furColor, bool isDomestic,
              bool isIndoor, string personality) 
        : base(name, age, weight, furColor, isDomestic)
    {
        IsIndoor = isIndoor;
        Personality = personality;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Домашний кот: {(IsIndoor ? "Да" : "Нет")}, Характер: {Personality}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }

    public Bird(string name, int age, double weight, double wingspan, bool canFly) 
        : base(name, age, weight)
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan}см, Умеет летать: {(CanFly ? "Да" : "Нет")}");
    }

    public virtual void Fly()
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} летит!");
        }
        else
        {
            Console.WriteLine($"{Name} не умеет летать");
        }
    }
}

// Производный класс Parrot (сложное наследование от Bird)
public class Parrot : Bird
{
    public string Vocabulary { get; set; }
    public string Color { get; set; }

    public Parrot(string name, int age, double weight, double wingspan, bool canFly,
                 string vocabulary, string color) 
        : base(name, age, weight, wingspan, canFly)
    {
        Vocabulary = vocabulary;
        Color = color;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Цвет: {Color}, Словарный запас: {Vocabulary}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: {Vocabulary}");
    }

    public void Talk()
    {
        Console.WriteLine($"{Name} повторяет: '{Vocabulary}'");
    }
}

// Использование классов
Console.WriteLine("=== Сложное наследование: Животные ===");

// Простое наследование
Mammal basicMammal = new Mammal("Млекопитающее", 3, 10.0, "Коричневый", true);
basicMammal.DisplayInfo();
basicMammal.FeedMilk();

Console.WriteLine();

// Сложное наследование - Dog наследует от Mammal
Dog myDog = new Dog("Барсик", 3, 12.5, "Коричневый", true, "Лабрадор", "Большой");
myDog.DisplayInfo();
myDog.MakeSound();
myDog.Fetch();

Console.WriteLine();

// Глубокое сложное наследование - WorkingDog наследует от Dog
WorkingDog policeDog = new WorkingDog("Рекс", 4, 15.0, "Черный", true, 
                                     "Немецкая овчарка", "Средний", "Полицейская собака", true);
policeDog.DisplayInfo();
policeDog.MakeSound();
policeDog.DoJob();

Console.WriteLine();

// Сложное наследование - Cat наследует от Mammal
Cat myCat = new Cat("Мурка", 2, 4.2, "Серый", true, true, "Игривый");
myCat.DisplayInfo();
myCat.MakeSound();
myCat.Purr();

Console.WriteLine();

// Сложное наследование - Parrot наследует от Bird
Parrot myParrot = new Parrot("Кеша", 1, 0.3, 25.5, true, "Привет! Как дела?", "Зеленый");
myParrot.DisplayInfo();
myParrot.MakeSound();
myParrot.Talk();
myParrot.Fly();
````
