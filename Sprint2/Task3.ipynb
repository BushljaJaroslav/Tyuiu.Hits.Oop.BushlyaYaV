```
public interface IBorrowable
{
    bool IsAvailable { get; set; }
    string Borrower { get; set; }
    void Borrow(string borrower);
    void Return();
    DateTime DueDate { get; set; }
}

public interface IDigital
{
    string FileFormat { get; set; }
    double FileSize { get; set; }
    void Download();
    void OpenInApp();
}

public interface IRare
{
    bool IsRare { get; set; }
    decimal Valuation { get; set; }
    void Appraise();
    void HandleWithCare();
}

// Базовый класс Book
public class Book
{
    // Основные атрибуты
    public string Title { get; set; }
    public string Author { get; set; }
    public int YearOfPublication { get; set; }
    
    // Новые атрибуты
    public string ISBN { get; set; }
    public string Publisher { get; set; }
    public int PageCount { get; set; }
    public string Language { get; set; }
    public decimal Price { get; set; }

    // Конструктор
    public Book(string title, string author, int yearOfPublication, string isbn, string publisher, int pageCount, string language, decimal price)
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
        ISBN = isbn;
        Publisher = publisher;
        PageCount = pageCount;
        Language = language;
        Price = price;
    }

    // Основные методы
    public virtual string GetInfo()
    {
        return $"Книга: '{Title}', Автор: {Author}, Год: {YearOfPublication}, ISBN: {ISBN}";
    }

    public virtual void Read()
    {
        Console.WriteLine($"Чтение книги: '{Title}' ({PageCount} страниц)");
    }

    public virtual void Borrow()
    {
        Console.WriteLine($"Книга '{Title}' выдана на чтение");
    }

    // Новые методы
    public virtual void Review()
    {
        Console.WriteLine($"Просмотр книги '{Title}' издательства {Publisher}");
    }

    public virtual decimal CalculateDiscount(int percent)
    {
        decimal discount = Price * percent / 100;
        Console.WriteLine($"Скидка на '{Title}': {discount:C}");
        return discount;
    }

    public void DisplayDetails()
    {
        Console.WriteLine($"Детали: {PageCount} стр., Язык: {Language}, Цена: {Price:C}");
    }
}

// Производный класс Textbook (Учебник) - простое наследование
public class Textbook : Book
{
    // Дополнительные атрибуты
    public string Subject { get; set; }
    public string GradeLevel { get; set; }
    public bool HasExercises { get; set; }
    public string Curriculum { get; set; }

    // Конструктор
    public Textbook(string title, string author, int yearOfPublication, string isbn, string publisher, int pageCount, string language, decimal price, string subject, string gradeLevel, bool hasExercises, string curriculum) 
        : base(title, author, yearOfPublication, isbn, publisher, pageCount, language, price)
    {
        Subject = subject;
        GradeLevel = gradeLevel;
        HasExercises = hasExercises;
        Curriculum = curriculum;
    }

    // Переопределение методов
    public override void Read()
    {
        Console.WriteLine($"Изучение учебника по предмету '{Subject}' для {GradeLevel}: '{Title}'");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}, Класс: {GradeLevel}";
    }

    // Новые методы
    public void SolveExercises()
    {
        if (HasExercises)
        {
            Console.WriteLine($"Решение упражнений в учебнике '{Title}'");
        }
        else
        {
            Console.WriteLine($"В учебнике '{Title}' нет упражнений");
        }
    }

    public void PrepareForExam()
    {
        Console.WriteLine($"Подготовка к экзамену по '{Subject}' с использованием учебника '{Title}'");
    }

    public override void Review()
    {
        Console.WriteLine($"Просмотр учебника '{Title}' по программе {Curriculum}");
    }
}

// Производный класс Fiction (Художественная литература) - простое наследование
public class Fiction : Book
{
    // Дополнительные атрибуты
    public string Genre { get; set; }
    public string MainCharacter { get; set; }
    public string Setting { get; set; }
    public bool IsBestseller { get; set; }

    // Конструктор
    public Fiction(string title, string author, int yearOfPublication, string isbn, string publisher, int pageCount, string language, decimal price, string genre, string mainCharacter, string setting, bool isBestseller) 
        : base(title, author, yearOfPublication, isbn, publisher, pageCount, language, price)
    {
        Genre = genre;
        MainCharacter = mainCharacter;
        Setting = setting;
        IsBestseller = isBestseller;
    }

    // Переопределение методов
    public override void Borrow()
    {
        Console.WriteLine($"Художественная книга в жанре '{Genre}': '{Title}' выдана для удовольствия чтения");
    }

    public override string GetInfo()
    {
        string bestsellerInfo = IsBestseller ? " (БЕСТСЕЛЛЕР)" : "";
        return base.GetInfo() + $", Жанр: {Genre}{bestsellerInfo}";
    }

    // Новые методы
    public void AnalyzePlot()
    {
        Console.WriteLine($"Анализ сюжета книги '{Title}' с главным героем {MainCharacter}");
    }

    public void DiscussWithFriends()
    {
        Console.WriteLine($"Обсуждение книги '{Title}' с друзьями");
    }

    public override decimal CalculateDiscount(int percent)
    {
        decimal baseDiscount = base.CalculateDiscount(percent);
        if (IsBestseller)
        {
            decimal extraDiscount = Price * 0.05m; // Дополнительная 5% скидка для бестселлеров
            Console.WriteLine($"Дополнительная скидка для бестселлера: {extraDiscount:C}");
            return baseDiscount + extraDiscount;
        }
        return baseDiscount;
    }
}

// Производный класс ScientificLiterature (Научная литература) - простое наследование
public class ScientificLiterature : Book
{
    // Дополнительные атрибуты
    public string FieldOfScience { get; set; }
    public string ResearchMethod { get; set; }
    public bool PeerReviewed { get; set; }
    public string DOI { get; set; }

    // Конструктор
    public ScientificLiterature(string title, string author, int yearOfPublication, string isbn, string publisher, int pageCount, string language, decimal price, string fieldOfScience, string researchMethod, bool peerReviewed, string doi) 
        : base(title, author, yearOfPublication, isbn, publisher, pageCount, language, price)
    {
        FieldOfScience = fieldOfScience;
        ResearchMethod = researchMethod;
        PeerReviewed = peerReviewed;
        DOI = doi;
    }

    // Переопределение методов
    public override string GetInfo()
    {
        string peerReviewInfo = PeerReviewed ? " (Рецензировано)" : "";
        return base.GetInfo() + $", Область науки: {FieldOfScience}{peerReviewInfo}";
    }

    public override void Read()
    {
        Console.WriteLine($"Исследование научной работы в области '{FieldOfScience}': '{Title}'");
    }

    // Новые методы
    public void CiteSource()
    {
        Console.WriteLine($"Цитирование научной работы: {Author} ({YearOfPublication}). {Title}. DOI: {DOI}");
    }

    public void ConductResearch()
    {
        Console.WriteLine($"Проведение исследования по методике: {ResearchMethod}");
    }

    public override void Review()
    {
        Console.WriteLine($"Рецензирование научной работы '{Title}' в области {FieldOfScience}");
    }
}

// Класс BorrowableBook - сложное наследование (наследует от Book и реализует IBorrowable)
public class BorrowableBook : Book, IBorrowable
{
    // Реализация IBorrowable
    public bool IsAvailable { get; set; }
    public string Borrower { get; set; }
    public DateTime DueDate { get; set; }
    
    // Новые атрибуты
    public string LibraryCode { get; set; }
    public int BorrowCount { get; set; }

    public BorrowableBook(string title, string author, int yearOfPublication, string isbn, string publisher, int pageCount, string language, decimal price, string libraryCode) 
        : base(title, author, yearOfPublication, isbn, publisher, pageCount, language, price)
    {
        LibraryCode = libraryCode;
        IsAvailable = true;
        BorrowCount = 0;
    }

    // Реализация методов IBorrowable
    public void Borrow(string borrower)
    {
        if (IsAvailable)
        {
            Borrower = borrower;
            IsAvailable = false;
            DueDate = DateTime.Now.AddDays(30);
            BorrowCount++;
            Console.WriteLine($"Книга '{Title}' выдана {borrower} до {DueDate:dd.MM.yyyy}");
        }
        else
        {
            Console.WriteLine($"Книга '{Title}' уже выдана {Borrower}");
        }
    }

    public void Return()
    {
        if (!IsAvailable)
        {
            Console.WriteLine($"Книга '{Title}' возвращена {Borrower}");
            Borrower = null;
            IsAvailable = true;
            DueDate = DateTime.MinValue;
        }
    }

    // Переопределение методов
    public override void Borrow()
    {
        Console.WriteLine($"Используйте метод Borrow(string borrower) для выдачи книги");
    }

    public void CheckOverdue()
    {
        if (!IsAvailable && DateTime.Now > DueDate)
        {
            Console.WriteLine($"КНИГА ПРОСРОЧЕНА! '{Title}' должна была быть возвращена {DueDate:dd.MM.yyyy}");
        }
    }

    public void DisplayBorrowHistory()
    {
        Console.WriteLine($"Книга '{Title}' выдавалась {BorrowCount} раз");
    }
}

// Класс RareBook - сложное наследование (наследует от Book и реализует IRare)
public class RareBook : Book, IRare
{
    // Реализация IRare
    public bool IsRare { get; set; }
    public decimal Valuation { get; set; }
    
    // Новые атрибуты
    public string Condition { get; set; }
    public int PrintYear { get; set; }
    public string Collector { get; set; }

    public RareBook(string title, string author, int yearOfPublication, string isbn, string publisher, int pageCount, string language, decimal price, string condition, int printYear, string collector) 
        : base(title, author, yearOfPublication, isbn, publisher, pageCount, language, price)
    {
        Condition = condition;
        PrintYear = printYear;
        Collector = collector;
        IsRare = true;
    }

    // Реализация методов IRare
    public void Appraise()
    {
        Valuation = Price * 10; // Редкие книги ценнее в 10 раз
        Console.WriteLine($"Оценка редкой книги '{Title}': {Valuation:C}");
    }

    public void HandleWithCare()
    {
        Console.WriteLine($"Осторожное обращение с редкой книгой '{Title}' в состоянии '{Condition}'");
    }

    // Переопределение методов
    public override string GetInfo()
    {
        return base.GetInfo() + $", Состояние: {Condition}, Год печати: {PrintYear}, Коллекционер: {Collector}";
    }

    public override void Read()
    {
        HandleWithCare();
        Console.WriteLine($"Чтение редкого издания '{Title}' в специальных условиях");
    }

    // Новые методы
    public void Restore()
    {
        Console.WriteLine($"Реставрация редкой книги '{Title}'");
    }

    public void Insure()
    {
        Console.WriteLine($"Страхование редкой книги '{Title}' на сумму {Valuation:C}");
    }
}

// Класс EBook - множественное наследование (наследует от Book и реализует IDigital, IBorrowable)
public class EBook : Book, IDigital, IBorrowable
{
    // Реализация IDigital
    public string FileFormat { get; set; }
    public double FileSize { get; set; }
    
    // Реализация IBorrowable
    public bool IsAvailable { get; set; }
    public string Borrower { get; set; }
    public DateTime DueDate { get; set; }
    
    // Новые атрибуты
    public string DownloadLink { get; set; }
    public bool DRMProtected { get; set; }

    public EBook(string title, string author, int yearOfPublication, string isbn, string publisher, int pageCount, string language, decimal price, string fileFormat, double fileSize, string downloadLink, bool drmProtected) 
        : base(title, author, yearOfPublication, isbn, publisher, pageCount, language, price)
    {
        FileFormat = fileFormat;
        FileSize = fileSize;
        DownloadLink = downloadLink;
        DRMProtected = drmProtected;
        IsAvailable = true;
    }

    // Реализация методов IDigital
    public void Download()
    {
        Console.WriteLine($"Скачивание электронной книги '{Title}' ({FileSize} MB) в формате {FileFormat}");
    }

    public void OpenInApp()
    {
        Console.WriteLine($"Открытие книги '{Title}' в читальном приложении");
    }

    // Реализация методов IBorrowable
    public void Borrow(string borrower)
    {
        Borrower = borrower;
        IsAvailable = false;
        DueDate = DateTime.Now.AddDays(14); // Электронные книги на 14 дней
        Console.WriteLine($"Электронная книга '{Title}' выдана {borrower} до {DueDate:dd.MM.yyyy}");
    }

    public void Return()
    {
        Console.WriteLine($"Электронная книга '{Title}' автоматически возвращена");
        Borrower = null;
        IsAvailable = true;
    }

    // Переопределение методов
    public override void Read()
    {
        OpenInApp();
        Console.WriteLine($"Чтение электронной книги '{Title}' на устройстве");
    }

    public override void Borrow()
    {
        Console.WriteLine($"Используйте метод Borrow(string borrower) для выдачи электронной книги");
    }

    // Новые методы
    public void SendToDevice(string device)
    {
        Console.WriteLine($"Отправка книги '{Title}' на устройство: {device}");
    }

    public void ConvertFormat(string newFormat)
    {
        Console.WriteLine($"Конвертация книги '{Title}' из {FileFormat} в {newFormat}");
        FileFormat = newFormat;
    }
}

// Класс EducationalTextbook - сложное наследование (наследует от Textbook и реализует IDigital, IBorrowable)
public class EducationalTextbook : Textbook, IDigital, IBorrowable
{
    // Реализация IDigital
    public string FileFormat { get; set; }
    public double FileSize { get; set; }
    
    // Реализация IBorrowable
    public bool IsAvailable { get; set; }
    public string Borrower { get; set; }
    public DateTime DueDate { get; set; }
    
    // Новые атрибуты
    public bool HasInteractiveContent { get; set; }
    public string OnlinePlatform { get; set; }

    public EducationalTextbook(string title, string author, int yearOfPublication, string isbn, string publisher, int pageCount, string language, decimal price, string subject, string gradeLevel, bool hasExercises, string curriculum, string fileFormat, double fileSize, bool hasInteractiveContent, string onlinePlatform) 
        : base(title, author, yearOfPublication, isbn, publisher, pageCount, language, price, subject, gradeLevel, hasExercises, curriculum)
    {
        FileFormat = fileFormat;
        FileSize = fileSize;
        HasInteractiveContent = hasInteractiveContent;
        OnlinePlatform = onlinePlatform;
        IsAvailable = true;
    }

    // Реализация методов IDigital
    public void Download()
    {
        Console.WriteLine($"Скачивание учебника '{Title}' по {Subject} для {GradeLevel}");
    }

    public void OpenInApp()
    {
        Console.WriteLine($"Открытие интерактивного учебника '{Title}' на платформе {OnlinePlatform}");
    }

    // Реализация методов IBorrowable
    public void Borrow(string borrower)
    {
        Borrower = borrower;
        IsAvailable = false;
        DueDate = DateTime.Now.AddDays(60); // Учебники на 60 дней
        Console.WriteLine($"Образовательный учебник '{Title}' выдан {borrower} до {DueDate:dd.MM.yyyy}");
    }

    public void Return()
    {
        Console.WriteLine($"Образовательный учебник '{Title}' возвращен");
        Borrower = null;
        IsAvailable = true;
    }

    // Переопределение методов
    public override void Read()
    {
        if (HasInteractiveContent)
        {
            OpenInApp();
            Console.WriteLine($"Изучение интерактивного учебника '{Title}' с упражнениями");
        }
        else
        {
            base.Read();
        }
    }

    // Новые методы
    public void TakeOnlineQuiz()
    {
        Console.WriteLine($"Прохождение онлайн-теста по учебнику '{Title}'");
    }

    public void AccessSupplementaryMaterials()
    {
        Console.WriteLine($"Доступ к дополнительным материалам учебника '{Title}'");
    }
}

// Демонстрация работы классов
public class LibraryDemo
{
    public static void Demonstrate()
    {
        Console.WriteLine("=== ДЕМОНСТРАЦИЯ РАБОТЫ БИБЛИОТЕКИ ===\n");

        // Создание объектов разных типов книг
        Book[] books = new Book[]
        {
            new Book("Основы программирования", "Иван Петров", 2020, "978-1234567890", "ТехноЛит", 350, "Русский", 1200m),
            new Textbook("Математика для начинающих", "Анна Сидорова", 2019, "978-0987654321", "Образование", 280, "Русский", 950m, "Математика", "5-6 класс", true, "ФГОС"),
            new Fiction("Преступление и наказание", "Федор Достоевский", 1866, "978-1122334455", "Классика", 672, "Русский", 650m, "Роман", "Родион Раскольников", "Санкт-Петербург", true),
            new ScientificLiterature("Теория относительности", "Альберт Эйнштейн", 1916, "978-5566778899", "Наука", 150, "Английский", 2300m, "Физика", "Теоретический анализ", true, "10.1000/182"),
            new BorrowableBook("Война и мир", "Лев Толстой", 1869, "978-9988776655", "Русская классика", 1225, "Русский", 1800m, "LIB-001"),
            new RareBook("Первое издание Евгения Онегина", "Александр Пушкин", 1833, "978-7744112233", "Императорская типография", 240, "Русский", 5000m, "Отличное", 1833, "Иванов И.И."),
            new EBook("Цифровая эра", "Петр Digital", 2023, "978-4455667788", "TechBooks", 320, "Русский", 800m, "PDF", 15.5, "https://library.ru/ebooks/123", false),
            new EducationalTextbook("Информатика 10-11", "Сергей Программистов", 2022, "978-8899001122", "Просвещение", 400, "Русский", 1100m, "Информатика", "10-11 класс", true, "ФГОС", "EPUB", 25.0, true, "Российская электронная школа")
        };

        // Демонстрация полиморфизма
        foreach (var book in books)
        {
            Console.WriteLine("=" .PadRight(50, '='));
            Console.WriteLine(book.GetInfo());
            book.Read();
            book.Borrow();
            book.Review();
            book.CalculateDiscount(10);
            book.DisplayDetails();
            Console.WriteLine();
        }

        // Демонстрация специфического поведения
        Console.WriteLine("\n" + "=" .PadRight(60, '='));
        Console.WriteLine("СПЕЦИФИЧЕСКОЕ ПОВЕДЕНИЕ И ИНТЕРФЕЙСЫ");
        Console.WriteLine("=" .PadRight(60, '='));

        // Работа с BorrowableBook
        BorrowableBook borrowable = (BorrowableBook)books[4];
        borrowable.Borrow("Иван Иванов");
        borrowable.CheckOverdue();
        borrowable.DisplayBorrowHistory();

        Console.WriteLine();

        // Работа с RareBook
        RareBook rare = (RareBook)books[5];
        rare.Appraise();
        rare.Insure();
        rare.HandleWithCare();

        Console.WriteLine();

        // Работа с EBook
        EBook ebook = (EBook)books[6];
        ebook.Download();
        ebook.Borrow("Петр Петров");
        ebook.SendToDevice("Планшет");

        Console.WriteLine();

        // Работа с EducationalTextbook
        EducationalTextbook eduTextbook = (EducationalTextbook)books[7];
        eduTextbook.Borrow("Студент группы ИСП-101");
        eduTextbook.TakeOnlineQuiz();
        eduTextbook.SolveExercises();
        eduTextbook.AccessSupplementaryMaterials();

        Console.WriteLine();

        // Демонстрация возврата книг
        Console.WriteLine("ВОЗВРАТ КНИГ:");
        borrowable.Return();
        ebook.Return();
        eduTextbook.Return();
    }
}

// Запуск демонстрации
LibraryDemo.Demonstrate();
```
