### Интерфейсы в C#

**Интерфейс** — это контракт, который определяет набор методов и свойств, которые класс должен реализовать. Интерфейсы позволяют создать гибкую и расширяемую архитектуру, поскольку они отделяют определение поведения от его реализации.

Давайте подробно разберем понятия интерфейсов и управления зависимостями в C# на примере классов `Vehicle`, `Car` и `Motorcycle`.

#### Пример интерфейса

Предположим, мы имеем интерфейс `IVehicle`, который определяет поведение транспортных средств. Ниже приведен пример его определения:

```csharp
public interface IVehicle
{
    void Start();            // Метод для запуска транспортного средства
    void Stop();             // Метод для остановки транспортного средства
    int GetNumberOfWheels(); // Метод для получения количества колес
}
```

Интерфейс `IVehicle` требует, чтобы любой класс, реализующий его, обеспечил реализацию этих методов.

### Реализация интерфейса

Теперь создадим классы `Car` и `Motorcycle`, которые будут реализовывать этот интерфейс:

```csharp
public class Car : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Машина запущена.");
    }

    public void Stop()
    {
        Console.WriteLine("Машина остановлена.");
    }

    public int GetNumberOfWheels()
    {
        return 4; // У машины 4 колеса
    }
}

public class Motorcycle : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Мотоцикл запущен.");
    }

    public void Stop()
    {
        Console.WriteLine("Мотоцикл остановлен.");
    }

    public int GetNumberOfWheels()
    {
        return 2; // У мотоцикла 2 колеса
    }
}
```

Каждый класс (`Car` и `Motorcycle`) реализует все методы, определенные в интерфейсе `IVehicle`, что позволяет создать единый интерфейс для работы с разными типами транспортных средств.

### Управление зависимостями (Dependency Injection)

**Управление зависимостями** — это паттерн проектирования, который помогает управлять созданием зависимостей в классе. Этот подход позволяет внедрять (или "вводить") зависимости через конструктор, методы или свойства, вместо того чтобы создавать их внутри класса. Это делает код более чистым и облегчает его тестирование.

#### Пример управления зависимостями

Создадим класс `VehicleService`, который будет зависеть от интерфейса `IVehicle`:

```csharp
public class VehicleService
{
    private readonly IVehicle _vehicle;

    // Внедрение зависимости через конструктор
    public VehicleService(IVehicle vehicle)
    {
        _vehicle = vehicle; 
    }

    public void UseVehicle()
    {
        _vehicle.Start(); // Запускаем транспортное средство
        Console.WriteLine($"У этого транспортного средства {_vehicle.GetNumberOfWheels()} колеса.");
        _vehicle.Stop(); // Останавливаем транспортное средство
    }
}
```

В этом примере класс `VehicleService` принимает в своем конструкторе объект, реализующий интерфейс `IVehicle`. Это позволяет `VehicleService` работать с любым объектом, который реализует `IVehicle`, обеспечивая гибкость и уменьшая связывание между классами.

### Использование классов

Теперь давайте создадим объекты `Car` и `Motorcycle` и используем наш сервис:

```csharp
IVehicle myCar = new Car(); // Создаем объект машины
VehicleService carService = new VehicleService(myCar); // Передаем машину в сервис
carService.UseVehicle(); // Используем сервис с машиной

IVehicle myMotorcycle = new Motorcycle(); // Создаем объект мотоцикла
VehicleService motorcycleService = new VehicleService(myMotorcycle); // Передаем мотоцикл в сервис
motorcycleService.UseVehicle(); // Используем сервис с мотоциклом

```

### Результат выполнения

При запуске данной программы будет следующий вывод:

```
Машина запущена.
У этого транспортного средства 4 колеса.
Машина остановлена.
Мотоцикл запущен.
У этого транспортного средства 2 колеса.
Мотоцикл остановлен.
```

### Заключение

Использование интерфейсов и управления зависимостями позволяет создать более организованный и тестируемый код. Интерфейсы определяют стандартные поведения, а управление зависимостями снижает связность между компонентами. Это делает систему более гибкой и упрощает процесс её поддержки и расширения в будущем.

````
public interface IVehicle
{
    void Start();
    void Stop();
    int GetNumberOfWheels();
}

// Класс Car
public class Car : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Машина запущена.");
    }

    public void Stop()
    {
        Console.WriteLine("Машина остановлена.");
    }

    public int GetNumberOfWheels()
    {
        return 4;
    }
}

// Класс Motorcycle
public class Motorcycle : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Мотоцикл запущен.");
    }

    public void Stop()
    {
        Console.WriteLine("Мотоцикл остановлен.");
    }

    public int GetNumberOfWheels()
    {
        return 2;
    }
}

// Сервис с внедрением зависимости
public class VehicleService
{
    private readonly IVehicle _vehicle;

    // Внедрение зависимости через конструктор
    public VehicleService(IVehicle vehicle)
    {
        _vehicle = vehicle;
    }

    public void UseVehicle()
    {
        _vehicle.Start();
        Console.WriteLine($"У этого транспортного средства {_vehicle.GetNumberOfWheels()} колеса.");
        _vehicle.Stop();
    }
}

// ПРОВЕРКА РАБОТЫ
Console.WriteLine("=== ПРОВЕРКА УПРАВЛЕНИЯ ЗАВИСИМОСТЯМИ ===\n");

// Создание объектов
IVehicle myCar = new Car();
IVehicle myMotorcycle = new Motorcycle();

// Использование сервиса с разными зависимостями
Console.WriteLine("1. РАБОТА С АВТОМОБИЛЕМ:");
VehicleService carService = new VehicleService(myCar);
carService.UseVehicle();

Console.WriteLine("\n2. РАБОТА С МОТОЦИКЛОМ:");
VehicleService motorcycleService = new VehicleService(myMotorcycle);
motorcycleService.UseVehicle();

Console.WriteLine("\n3. ПОЛИМОРФИЗМ ЧЕРЕЗ СЕРВИС:");
IVehicle[] vehicles = new IVehicle[] { new Car(), new Motorcycle() };

foreach (IVehicle vehicle in vehicles)
{
    VehicleService service = new VehicleService(vehicle);
    service.UseVehicle();
    Console.WriteLine("---");
}
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также реализовать управление зависимостями.

````
public interface IAnimal
{
    string Name { get; set; }
    void MakeSound();
    void Move();
    string GetInfo();
}

// Класс Dog
public class Dog : IAnimal
{
    public string Name { get; set; }
    public string Breed { get; set; }

    public Dog(string name, string breed)
    {
        Name = name;
        Breed = breed;
    }

    public void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }

    public void Move()
    {
        Console.WriteLine($"{Name} бежит виляя хвостом");
    }

    public string GetInfo()
    {
        return $"Собака {Name}, порода {Breed}";
    }

    // Специфичный метод
    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит палку!");
    }
}

// Класс Cat
public class Cat : IAnimal
{
    public string Name { get; set; }
    public string Color { get; set; }

    public Cat(string name, string color)
    {
        Name = name;
        Color = color;
    }

    public void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    public void Move()
    {
        Console.WriteLine($"{Name} грациозно крадется");
    }

    public string GetInfo()
    {
        return $"Кот {Name}, цвет {Color}";
    }

    // Специфичный метод
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }
}

// Класс Bird
public class Bird : IAnimal
{
    public string Name { get; set; }
    public double Wingspan { get; set; }

    public Bird(string name, double wingspan)
    {
        Name = name;
        Wingspan = wingspan;
    }

    public void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    public void Move()
    {
        Console.WriteLine($"{Name} летает с размахом крыльев {Wingspan} см");
    }

    public string GetInfo()
    {
        return $"Птица {Name}, размах крыльев {Wingspan} см";
    }

    // Специфичный метод
    public void BuildNest()
    {
        Console.WriteLine($"{Name} строит гнездо");
    }
}

// Класс Fish
public class Fish : IAnimal
{
    public string Name { get; set; }
    public string WaterType { get; set; }

    public Fish(string name, string waterType)
    {
        Name = name;
        WaterType = waterType;
    }

    public void MakeSound()
    {
        Console.WriteLine($"{Name} издает булькающие звуки");
    }

    public void Move()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде");
    }

    public string GetInfo()
    {
        return $"Рыба {Name}, тип воды {WaterType}";
    }

    // Специфичный метод
    public void Swim()
    {
        Console.WriteLine($"{Name} плывет в {WaterType} воде");
    }
}

// Сервис для ухода за животными (внедрение зависимости)
public class AnimalCareService
{
    private readonly IAnimal _animal;

    // Внедрение зависимости через конструктор
    public AnimalCareService(IAnimal animal)
    {
        _animal = animal;
    }

    public void PerformDailyCare()
    {
        Console.WriteLine($"\n=== УХОД ЗА {_animal.Name.ToUpper()} ===");
        Console.WriteLine(_animal.GetInfo());
        _animal.MakeSound();
        _animal.Move();
        Console.WriteLine("Уход завершен");
    }

    public void FeedAnimal(string food)
    {
        Console.WriteLine($"Кормление {_animal.Name}: {food}");
    }

    public void PlayWithAnimal()
    {
        Console.WriteLine($"Игра с {_animal.Name}");
        
        // Проверка типа для специфичных методов
        if (_animal is Dog dog)
        {
            dog.Fetch();
        }
        else if (_animal is Cat cat)
        {
            cat.Purr();
        }
        else if (_animal is Bird bird)
        {
            bird.BuildNest();
        }
        else if (_animal is Fish fish)
        {
            fish.Swim();
        }
    }
}

// Сервис для ветеринара
public class VeterinarianService
{
    private readonly IAnimal _animal;

    // Внедрение зависимости через конструктор
    public VeterinarianService(IAnimal animal)
    {
        _animal = animal;
    }

    public void PerformHealthCheck()
    {
        Console.WriteLine($"\n=== МЕДИЦИНСКИЙ ОСМОТР {_animal.Name.ToUpper()} ===");
        Console.WriteLine($"Осмотр: {_animal.GetInfo()}");
        _animal.MakeSound();
        Console.WriteLine("Здоровье в норме");
    }

    public void Vaccinate()
    {
        Console.WriteLine($"{_animal.Name} вакцинирован");
    }
}

// Фабрика для создания сервисов
public class AnimalServiceFactory
{
    public AnimalCareService CreateCareService(IAnimal animal)
    {
        return new AnimalCareService(animal);
    }

    public VeterinarianService CreateVetService(IAnimal animal)
    {
        return new VeterinarianService(animal);
    }
}

// ДЕМОНСТРАЦИЯ РАБОТЫ
Console.WriteLine("=== ДЕМОНСТРАЦИЯ УПРАВЛЕНИЯ ЗАВИСИМОСТЯМИ ===\n");

// Создание животных
IAnimal dog = new Dog("Шарик", "Овчарка");
IAnimal cat = new Cat("Барсик", "рыжий");
IAnimal bird = new Bird("Кеша", 15.5);
IAnimal fish = new Fish("Немо", "морская");

// Создание фабрики сервисов
AnimalServiceFactory factory = new AnimalServiceFactory();

// 1. Работа с собакой через сервисы
Console.WriteLine("1. УХОД ЗА СОБАКОЙ:");
AnimalCareService dogCareService = factory.CreateCareService(dog);
VeterinarianService dogVetService = factory.CreateVetService(dog);

dogCareService.PerformDailyCare();
dogCareService.FeedAnimal("мясо");
dogCareService.PlayWithAnimal();
dogVetService.PerformHealthCheck();
dogVetService.Vaccinate();

// 2. Работа с котом через сервисы
Console.WriteLine("\n2. УХОД ЗА КОТОМ:");
AnimalCareService catCareService = factory.CreateCareService(cat);
VeterinarianService catVetService = factory.CreateVetService(cat);

catCareService.PerformDailyCare();
catCareService.FeedAnimal("рыба");
catCareService.PlayWithAnimal();
catVetService.PerformHealthCheck();

// 3. Работа с птицей через сервисы
Console.WriteLine("\n3. УХОД ЗА ПТИЦЕЙ:");
AnimalCareService birdCareService = new AnimalCareService(bird);
birdCareService.PerformDailyCare();
birdCareService.FeedAnimal("зерно");
birdCareService.PlayWithAnimal();

// 4. Работа с рыбой через сервисы
Console.WriteLine("\n4. УХОД ЗА РЫБОЙ:");
AnimalCareService fishCareService = new AnimalCareService(fish);
fishCareService.PerformDailyCare();
fishCareService.FeedAnimal("планктон");
fishCareService.PlayWithAnimal();

// 5. Полиморфизм через сервисы
Console.WriteLine("\n5. ГРУППОВОЙ УХОД ЗА ВСЕМИ ЖИВОТНЫМИ:");
List<IAnimal> allAnimals = new List<IAnimal> { dog, cat, bird, fish };

foreach (IAnimal animal in allAnimals)
{
    AnimalCareService service = factory.CreateCareService(animal);
    service.PerformDailyCare();
    service.FeedAnimal("специальный корм");
    Console.WriteLine("---");
}

// 6. Демонстрация заменяемости зависимостей
Console.WriteLine("\n6. ЗАМЕНА ЗАВИСИМОСТЕЙ:");
Console.WriteLine("Тот же сервис ухода работает с разными животными:");

AnimalCareService flexibleService = new AnimalCareService(dog);
flexibleService.PerformDailyCare();

flexibleService = new AnimalCareService(cat);
flexibleService.PerformDailyCare();

// 7. Тестирование (имитация)
Console.WriteLine("\n7. ТЕСТИРОВАНИЕ С MOCK-ОБЪЕКТОМ:");

// Mock-объект для тестирования
public class TestAnimal : IAnimal
{
    public string Name { get; set; } = "Тестовое животное";
    
    public void MakeSound()
    {
        Console.WriteLine("Тестовый звук");
    }
    
    public void Move()
    {
        Console.WriteLine("Тестовое движение");
    }
    
    public string GetInfo()
    {
        return "Тестовая информация";
    }
}

IAnimal testAnimal = new TestAnimal();
AnimalCareService testService = new AnimalCareService(testAnimal);
testService.PerformDailyCare();
````
