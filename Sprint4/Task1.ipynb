<h1 style="color:DodgerBlue">Явная реализация интерфейса</h1>

В C# интерфейсы могут быть реализованы явным и неявным образом. Давайте подробно рассмотрим, что такое явная реализация интерфейсов, реализация интерфейсов в базовых и производных классах, а также наследование интерфейсов, используя примеры классов `Vehicle`, `Car` и `Motorcycle`.

### Явная реализация интерфейсов

Явная реализация интерфейса используется для того, чтобы скрыть методы интерфейса от стандартного контекста класса. Это означает, что методы интерфейса могут быть вызваны только через ссылку на интерфейс, а не через экземпляр класса.

#### Пример явной реализации

Рассмотрим интерфейс `IVehicle` и его явную реализацию в классе `Car`.

```csharp
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    // Явная реализация интерфейса
    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    // Неявная реализация метода
    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

В этом примере методы `Start` и `Stop` реализованы явно, и для их вызова необходимо использовать объект типа `IVehicle`:

```csharp

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry" };
myCar.Start(); // Корректно, вызывает явную реализацию
// myCar.Stop(); // Корректно, вызывает явную реализацию
```

Если вы попытаетесь вызвать `Start` или `Stop` непосредственно через экземпляр `Car`, это приведёт к ошибке компиляции:

```csharp
Car myCar = new Car();
myCar.Start(); // Ошибка компиляции
```

### Реализация интерфейсов в базовых и производных классах

Интерфейсы могут быть реализованы в базовых классах и наследоваться в производных классах. Давайте рассмотрим эту концепцию на примере, включающем классы `Vehicle`, `Car` и `Motorcycle`.

#### Базовый класс Vehicle

```csharp
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();  // Оставляем абстрактным
    public abstract void Stop();   // Оставляем абстрактным

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

#### Реализация в производных классах

Теперь реализуем `Car` и `Motorcycle`.

```csharp
public class Car : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }
}
```

Теперь оба класса `Car` и `Motorcycle` обязаны реализовать методы `Start` и `Stop`, которые определены в абстрактном классе `Vehicle`.

### Наследование интерфейсов

Интерфейсы могут наследовать друг от друга. Это позволяет создавать сложные иерархии интерфейсов. Например, мы можем создать интерфейс `IMotorizedVehicle`, который наследует `IVehicle`.

```csharp
public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}
```

Таким образом, `Motorcycle` теперь может реализовать все методы из обоих интерфейсов `IVehicle` и `IMotorizedVehicle`.

### Заключение

1. **Явная реализация** интерфейсов скрывает функциональность классов от общего доступа, что помогает избежать путаницы с методами.
   
2. **Реализация интерфейсов в базовых и производных классах** позволяет создавать иерархию классов и обеспечивает обязательность реализации методов в производных классах.

3. **Наследование интерфейсов** позволяет комбинировать несколько интерфейсов, создавая более сложные структуры, что делает код более гибким и расширяемым.

Эти механизмы являются важными инструментами в инструменте разработчика, позволяя создавать более структурированные и управляемые программы.

````
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

// Наследуемый интерфейс
public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

// Абстрактный базовый класс
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();
    public abstract void Stop();

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}

// Класс Car с явной реализацией интерфейса
public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    // Явная реализация интерфейса
    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен через явную реализацию.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен через явную реализацию.");
    }

    // Неявная реализация метода
    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }

    // Собственный метод класса
    public void Drive()
    {
        Console.WriteLine($"{Make} {Model} едет.");
    }
}

// Класс Motorcycle с наследованием интерфейсов
public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}

// ПРОВЕРКА РАБОТЫ
Console.WriteLine("=== ПРОВЕРКА ЯВНОЙ РЕАЛИЗАЦИИ ИНТЕРФЕЙСОВ ===\n");

// Проверка явной реализации в Car
Console.WriteLine("1. ЯВНАЯ РЕАЛИЗАЦИЯ В CAR:");
IVehicle carAsInterface = new Car { Make = "Toyota", Model = "Camry" };
carAsInterface.Start(); // Работает через интерфейс
carAsInterface.Stop();  // Работает через интерфейс
Console.WriteLine($"Скорость: {carAsInterface.GetCurrentSpeed()} км/ч");

// Попытка вызова через класс (не скомпилируется)
Car carAsClass = new Car { Make = "Honda", Model = "Civic" };
// carAsClass.Start(); // ОШИБКА КОМПИЛЯЦИИ - метод недоступен
carAsClass.Drive(); // Собственный метод работает
Console.WriteLine($"Скорость через класс: {carAsClass.GetCurrentSpeed()} км/ч");

Console.WriteLine("\n2. НАСЛЕДОВАНИЕ ИНТЕРФЕЙСОВ:");
IMotorizedVehicle motorcycle = new Motorcycle { Make = "Harley", Model = "Davidson" };
motorcycle.Start();    // Из IVehicle
motorcycle.Stop();     // Из IVehicle  
motorcycle.RevEngine(); // Из IMotorizedVehicle
Console.WriteLine($"Скорость мотоцикла: {motorcycle.GetCurrentSpeed()} км/ч");

Console.WriteLine("\n3. ПОЛИМОРФИЗМ ЧЕРЕЗ ИНТЕРФЕЙС:");
IVehicle[] vehicles = new IVehicle[]
{
    new Car { Make = "Ford", Model = "Focus" },
    new Motorcycle { Make = "Yamaha", Model = "R1" }
};

foreach (IVehicle vehicle in vehicles)
{
    vehicle.Start();
    Console.WriteLine($"Тип: {vehicle.GetType().Name}, Скорость: {vehicle.GetCurrentSpeed()} км/ч");
    vehicle.Stop();
    Console.WriteLine("---");
}
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также применить явнаю реализацию интерфейса.

````
public interface IAnimal
{
    void MakeSound();
    void Move();
    string GetInfo();
}

// Специализированный интерфейс для домашних животных
public interface IDomesticAnimal : IAnimal
{
    void PlayWithOwner();
}

// Абстрактный базовый класс
public abstract class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    protected double Weight { get; set; }

    protected Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    // Общий метод для всех животных
    public void Sleep()
    {
        Console.WriteLine($"{Name} спит.");
    }
}

// Класс Dog с явной реализацией интерфейса
public class Dog : IAnimal, IDomesticAnimal
{
    public string Name { get; set; }
    public string Breed { get; set; }
    private int energyLevel;

    public Dog(string name, string breed)
    {
        Name = name;
        Breed = breed;
        energyLevel = 100;
    }

    // Явная реализация базового интерфейса
    void IAnimal.MakeSound()
    {
        Console.WriteLine($"{Name} громко лает: ГАВ-ГАВ!");
    }

    void IAnimal.Move()
    {
        Console.WriteLine($"{Name} быстро бегает.");
        energyLevel -= 20;
    }

    string IAnimal.GetInfo()
    {
        return $"Собака {Name}, порода {Breed}, энергия: {energyLevel}%";
    }

    // Явная реализация наследуемого интерфейса
    void IDomesticAnimal.PlayWithOwner()
    {
        Console.WriteLine($"{Name} играет с хозяином и виляет хвостом!");
        energyLevel -= 15;
    }

    // Неявная реализация (доступна через класс)
    public void MakeSound()
    {
        Console.WriteLine($"{Name} тихо лает: гав-гав");
    }

    // Собственные методы класса
    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит палку!");
        energyLevel -= 10;
    }

    public void DisplayEnergy()
    {
        Console.WriteLine($"Уровень энергии {Name}: {energyLevel}%");
    }
}

// Класс Cat с обычной реализацией
public class Cat : Animal, IAnimal
{
    public string Color { get; set; }

    public Cat(string name, int age, double weight, string color) : base(name, age, weight)
    {
        Color = color;
    }

    // Неявная реализация интерфейса
    public void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    public void Move()
    {
        Console.WriteLine($"{Name} грациозно крадется.");
    }

    public string GetInfo()
    {
        return $"Кот {Name}, цвет {Color}, возраст {Age} лет, вес {Weight} кг";
    }

    // Собственный метод
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: мрррр...");
    }
}

// Класс Bird с множественным наследованием интерфейсов
public class Bird : Animal, IAnimal, IDomesticAnimal
{
    public double Wingspan { get; set; }

    public Bird(string name, int age, double weight, double wingspan) : base(name, age, weight)
    {
        Wingspan = wingspan;
    }

    public void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    public void Move()
    {
        Console.WriteLine($"{Name} летает с размахом крыльев {Wingspan} см.");
    }

    public string GetInfo()
    {
        return $"Птица {Name}, размах крыльев {Wingspan} см, возраст {Age} лет";
    }

    public void PlayWithOwner()
    {
        Console.WriteLine($"{Name} садится на плечо хозяину.");
    }

    // Собственный метод
    public void BuildNest()
    {
        Console.WriteLine($"{Name} строит гнездо.");
    }
}

// ДЕМОНСТРАЦИЯ РАБОТЫ
Console.WriteLine("=== ДЕМОНСТРАЦИЯ ЯВНОЙ РЕАЛИЗАЦИИ ИНТЕРФЕЙСОВ ===\n");

// 1. Явная реализация в Dog
Console.WriteLine("1. ЯВНАЯ РЕАЛИЗАЦИЯ В DOG:\n");

Dog dogAsClass = new Dog("Шарик", "Овчарка");
dogAsClass.MakeSound(); // Вызывает неявную реализацию
dogAsClass.Fetch();
dogAsClass.DisplayEnergy();

Console.WriteLine("\n--- Через интерфейс IAnimal ---");
IAnimal dogAsAnimal = new Dog("Рекс", "Доберман");
dogAsAnimal.MakeSound(); // Вызывает явную реализацию
dogAsAnimal.Move();
Console.WriteLine(dogAsAnimal.GetInfo());

Console.WriteLine("\n--- Через интерфейс IDomesticAnimal ---");
IDomesticAnimal dogAsDomestic = new Dog("Бобик", "Такса");
dogAsDomestic.MakeSound();
dogAsDomestic.PlayWithOwner();
dogAsDomestic.Move();

// 2. Обычная реализация в Cat
Console.WriteLine("\n2. ОБЫЧНАЯ РЕАЛИЗАЦИЯ В CAT:\n");
Cat cat = new Cat("Барсик", 2, 4.5, "рыжий");
cat.MakeSound();
cat.Move();
cat.Purr();
cat.Sleep();
Console.WriteLine(cat.GetInfo());

// 3. Множественное наследование интерфейсов в Bird
Console.WriteLine("\n3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ИНТЕРФЕЙСОВ В BIRD:\n");
Bird bird = new Bird("Кеша", 1, 0.3, 15.5);
bird.MakeSound();
bird.Move();
bird.PlayWithOwner();
bird.BuildNest();
Console.WriteLine(bird.GetInfo());

// 4. Полиморфизм через интерфейсы
Console.WriteLine("\n4. ПОЛИМОРФИЗМ ЧЕРЕЗ ИНТЕРФЕЙСЫ:\n");
List<IAnimal> animals = new List<IAnimal>
{
    new Dog("Дружок", "Лабрадор"),
    new Cat("Мурка", 3, 3.8, "серая"),
    new Bird("Чижик", 2, 0.2, 12.0)
};

foreach (IAnimal animal in animals)
{
    Console.WriteLine($"--- {animal.GetType().Name} ---");
    animal.MakeSound();
    animal.Move();
    Console.WriteLine(animal.GetInfo());
    
    // Проверка на специализированный интерфейс
    if (animal is IDomesticAnimal domestic)
    {
        domestic.PlayWithOwner();
    }
    Console.WriteLine();
}

// 5. Демонстрация ограничений явной реализации
Console.WriteLine("5. ОГРАНИЧЕНИЯ ЯВНОЙ РЕАЛИЗАЦИИ:");
Dog myDog = new Dog("Тузик", "Дворняжка");
// myDog.PlayWithOwner(); // ОШИБКА - метод доступен только через интерфейс

IDomesticAnimal domesticDog = myDog;
domesticDog.PlayWithOwner(); // Работает через интерфейс
````
