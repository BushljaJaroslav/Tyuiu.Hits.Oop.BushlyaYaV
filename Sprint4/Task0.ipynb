<h1 style="color:DodgerBlue">Определение и применение интерфейса</h1>



Интерфейсы в C# — это контракты, которые определяют набор методов и свойств, которые класс должен реализовать, если он "подписывается" на этот интерфейс. Они не содержат реализации, а лишь описывают, какие методы и свойства должны быть у класса. Интерфейсы позволяют создавать гибкую и расширяемую архитектуру, так как разные классы могут реализовывать один и тот же интерфейс, предоставляя свою индивидуальную логику.

### Пример интерфейса

Рассмотрим интерфейс `IVehicle`, который будет использоваться для классов `Vehicle`, `Car`, и `Motorcycle`.

```csharp
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}
```

### Реализация интерфейса в классах

Теперь создадим абстрактный класс `Vehicle`, который будет предоставлять базовые свойства, а потом создадим классы `Car` и `Motorcycle`, которые будут реализовывать интерфейс `IVehicle`.

```csharp
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public virtual void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    public virtual void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

### Реализация классов Car и Motorcycle

Теперь реализуем классы `Car` и `Motorcycle`, которые наследуются от `Vehicle` и могут добавлять свою специфическую логику.

```csharp
public class Car : Vehicle
{
    public int NumberOfDoors { get; set; }

    public override void Start()
    {
        base.Start();
        Console.WriteLine("Автомобиль готов к поездке.");
    }

    public override void Stop()
    {
        base.Stop();
        Console.WriteLine("Автомобиль полностью остановлен.");
    }

    public void Accelerate(int amount)
    {
        currentSpeed += amount;
        Console.WriteLine($"Автомобиль разгоняется до {currentSpeed} км/ч.");
    }
}

public class Motorcycle : Vehicle
{
    public bool HasSidecar { get; set; }

    public override void Start()
    {
        base.Start();
        Console.WriteLine("Мотоцикл готов к поездке.");
    }

    public override void Stop()
    {
        base.Stop();
        Console.WriteLine("Мотоцикл полностью остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}
```

### Использование классов

Теперь мы можем создать экземпляры классов `Car` и `Motorcycle` и использовать методы, определенные в интерфейсе:

```csharp

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry", NumberOfDoors = 4 };
myCar.Start();
((Car)myCar).Accelerate(50);
Console.WriteLine($"Текущая скорость: {myCar.GetCurrentSpeed()} км/ч");
        
IVehicle myMotorcycle = new Motorcycle { Make = "Harley-Davidson", Model = "Street", HasSidecar = false };
myMotorcycle.Start();
((Motorcycle)myMotorcycle).RevEngine();
myMotorcycle.Stop();
```

### Пояснения

1. **Интерфейс (IVehicle)**: Определяет методы, которые должны быть реализованы в классах, использующих этот интерфейс.
2. **Абстрактный класс (Vehicle)**: Реализует часть общей логики (например, свойства `Make`, `Model` и методы `Start`, `Stop`). Это позволяет избежать дублирования кода.
3. **Наследование и Полиморфизм**: Классы `Car` и `Motorcycle` наследуют `Vehicle` и могут расширять поведение класса, переопределяя методы и добавляя новые.
4. **Гибкость**: Используя интерфейсы, можно легко добавить другие транспортные средства, такие как `Truck`, просто реализовав тот же интерфейс.

Это создаёт хорошо структурированную и масштабируемую архитектуру, позволяя легко добавлять новые типы транспортных средств без изменения существующего кода. 

### Заключение
Интерфейсы в C# предоставляют мощный механизм для определения контрактов и достижения полиморфизма. Они позволяют разработчикам создавать гибкие и расширяемые системы, где классы могут быть легко заменены или добавлены без изменения существующего кода, который использует эти интерфейсы.

````
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

// Абстрактный класс Vehicle
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public virtual void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    public virtual void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}

// Класс Car
public class Car : Vehicle
{
    public int NumberOfDoors { get; set; }

    public override void Start()
    {
        base.Start();
        Console.WriteLine("Автомобиль готов к поездке.");
    }

    public override void Stop()
    {
        base.Stop();
        Console.WriteLine("Автомобиль полностью остановлен.");
    }

    public void Accelerate(int amount)
    {
        currentSpeed += amount;
        Console.WriteLine($"Автомобиль разгоняется до {currentSpeed} км/ч.");
    }
}

// Класс Motorcycle
public class Motorcycle : Vehicle
{
    public bool HasSidecar { get; set; }

    public override void Start()
    {
        base.Start();
        Console.WriteLine("Мотоцикл готов к поездке.");
    }

    public override void Stop()
    {
        base.Stop();
        Console.WriteLine("Мотоцикл полностью остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}

Console.WriteLine("=== ПРОВЕРКА РАБОТЫ ИНТЕРФЕЙСА И АБСТРАКТНОГО КЛАССА ===\n");

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry", NumberOfDoors = 4 };
IVehicle myMotorcycle = new Motorcycle { Make = "Harley-Davidson", Model = "Street", HasSidecar = false };

Console.WriteLine("=== ИСПОЛЬЗОВАНИЕ МЕТОДОВ ИНТЕРФЕЙСА IVEHICLE ===\n");

myCar.Start();
((Car)myCar).Accelerate(50);
Console.WriteLine($"Текущая скорость автомобиля: {myCar.GetCurrentSpeed()} км/ч");
myCar.Stop();

Console.WriteLine();

myMotorcycle.Start();
((Motorcycle)myMotorcycle).RevEngine();
Console.WriteLine($"Текущая скорость мотоцикла: {myMotorcycle.GetCurrentSpeed()} км/ч");
myMotorcycle.Stop();

// Демонстрация полиморфизма
Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ===\n");

IVehicle[] vehicles = new IVehicle[]
{
    new Car { Make = "Honda", Model = "Civic", NumberOfDoors = 4 },
    new Motorcycle { Make = "Yamaha", Model = "YZF-R3", HasSidecar = false },
    new Car { Make = "Ford", Model = "Focus", NumberOfDoors = 2 }
};

foreach (IVehicle vehicle in vehicles)
{
    vehicle.Start();
    Console.WriteLine($"Текущая скорость: {vehicle.GetCurrentSpeed()} км/ч");
    vehicle.Stop();
    Console.WriteLine("---");
}
````

````


// Интерфейс для животных
public interface IAnimal
{
    string Name { get; set; }
    int Age { get; set; }
    void MakeSound();
    void Move();
    void Eat(string food);
}

// Абстрактный базовый класс Animal
public abstract class Animal : IAnimal
{
    // Свойства с геттерами и сеттерами
    public string Name { get; set; }
    public int Age { get; set; }
    protected double Weight { get; set; }
    protected int Energy { get; set; }

    // Конструктор
    protected Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
        Energy = 100;
    }

    // Виртуальные методы с базовой реализацией
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук");
        Energy -= 5;
    }

    public virtual void Move()
    {
        Console.WriteLine($"{Name} перемещается");
        Energy -= 10;
    }

    public virtual void Eat(string food)
    {
        Console.WriteLine($"{Name} ест {food}");
        Energy += 20;
        Weight += 0.1;
    }

    // Абстрактный метод - должен быть реализован в производных классах
    public abstract void DisplayInfo();

    // Защищенный метод для внутреннего использования
    protected void Sleep()
    {
        Console.WriteLine($"{Name} спит и восстанавливает энергию");
        Energy += 30;
    }

    public void CheckEnergy()
    {
        if (Energy < 30)
        {
            Console.WriteLine($"{Name} устал и нуждается в отдыхе");
        }
    }
}

// Производный класс Dog
public class Dog : Animal
{
    public string Breed { get; set; }
    public string Owner { get; set; }

    public Dog(string name, int age, double weight, string breed, string owner) 
        : base(name, age, weight)
    {
        Breed = breed;
        Owner = owner;
    }

    // Реализация абстрактного метода
    public override void DisplayInfo()
    {
        Console.WriteLine($"Собака: {Name}, Порода: {Breed}, Возраст: {Age} лет, Вес: {Weight} кг, Владелец: {Owner}");
    }

    // Переопределение виртуальных методов
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
        Energy -= 3;
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} бежит виляя хвостом");
        Energy -= 8;
    }

    public override void Eat(string food)
    {
        if (food == "мясо")
        {
            Console.WriteLine($"{Name} с удовольствием ест {food}");
            Energy += 25;
        }
        else
        {
            base.Eat(food);
        }
    }

    // Специфичный метод для собак
    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит палку!");
        Energy -= 15;
    }
}

// Производный класс Cat
public class Cat : Animal
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }

    public Cat(string name, int age, double weight, string color, bool isIndoor) 
        : base(name, age, weight)
    {
        Color = color;
        IsIndoor = isIndoor;
    }

    // Реализация абстрактного метода
    public override void DisplayInfo()
    {
        string indoorStatus = IsIndoor ? "домашний" : "уличный";
        Console.WriteLine($"Кот: {Name}, Цвет: {Color}, Возраст: {Age} лет, Вес: {Weight} кг, {indoorStatus}");
    }

    // Переопределение виртуальных методов
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
        Energy -= 2;
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} грациозно крадется");
        Energy -= 5;
    }

    public override void Eat(string food)
    {
        if (food == "рыба")
        {
            Console.WriteLine($"{Name} с удовольствием ест {food}");
            Energy += 30;
        }
        else
        {
            base.Eat(food);
        }
    }

    // Специфичный метод для котов
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }

    public Bird(string name, int age, double weight, double wingspan, bool canFly) 
        : base(name, age, weight)
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }

    // Реализация абстрактного метода
    public override void DisplayInfo()
    {
        string flyAbility = CanFly ? "летающая" : "нелетающая";
        Console.WriteLine($"Птица: {Name}, Размах крыльев: {Wingspan} см, Возраст: {Age} лет, Вес: {Weight} кг, {flyAbility}");
    }

    // Переопределение виртуальных методов
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
        Energy -= 4;
    }

    public override void Move()
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} летит по небу");
            Energy -= 12;
        }
        else
        {
            Console.WriteLine($"{Name} прыгает по земле");
            Energy -= 6;
        }
    }

    public override void Eat(string food)
    {
        if (food == "зерно" || food == "семечки")
        {
            Console.WriteLine($"{Name} клюет {food}");
            Energy += 15;
        }
        else
        {
            base.Eat(food);
        }
    }

    // Специфичный метод для птиц
    public void BuildNest()
    {
        Console.WriteLine($"{Name} строит гнездо");
        Energy -= 20;
    }
}

// Производный класс Fish
public class Fish : Animal
{
    public string WaterType { get; set; }
    public double SwimSpeed { get; set; }

    public Fish(string name, int age, double weight, string waterType, double swimSpeed) 
        : base(name, age, weight)
    {
        WaterType = waterType;
        SwimSpeed = swimSpeed;
    }

    // Реализация абстрактного метода
    public override void DisplayInfo()
    {
        Console.WriteLine($"Рыба: {Name}, Тип воды: {WaterType}, Возраст: {Age} лет, Вес: {Weight} кг, Скорость плавания: {SwimSpeed} км/ч");
    }

    // Переопределение виртуальных методов
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} издает булькающие звуки");
        Energy -= 1;
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде");
        Energy -= 8;
    }

    public override void Eat(string food)
    {
        if (food == "планктон" || food == "водоросли")
        {
            Console.WriteLine($"{Name} фильтрует {food}");
            Energy += 18;
        }
        else
        {
            base.Eat(food);
        }
    }

    // Специфичный метод для рыб
    public void Swim()
    {
        Console.WriteLine($"{Name} плывет со скоростью {SwimSpeed} км/ч");
        Energy -= 10;
    }
}

// Демонстрация работы с интерфейсами и абстрактными классами
Console.WriteLine("=== ДЕМОНСТРАЦИЯ ИНТЕРФЕЙСОВ И АБСТРАКТНЫХ КЛАССОВ ===\n");

// Создаем список животных через интерфейс IAnimal
List<IAnimal> animals = new List<IAnimal>
{
    new Dog("Шарик", 3, 12.5, "Овчарка", "Иван"),
    new Cat("Барсик", 2, 4.2, "рыжий", true),
    new Bird("Кеша", 1, 0.3, 15.5, true),
    new Fish("Немо", 1, 0.5, "морская", 2.5)
};

// Демонстрация полиморфизма через интерфейс
Console.WriteLine("=== ПОЛИМОРФИЗМ ЧЕРЕЗ ИНТЕРФЕЙС IANIMAL ===\n");
foreach (IAnimal animal in animals)
{
    animal.MakeSound();
    animal.Move();
    animal.Eat("еда");
    Console.WriteLine();
}

// Работа с конкретными типами
Console.WriteLine("=== РАБОТА С КОНКРЕТНЫМИ ТИПАМИ ===\n");
foreach (IAnimal animal in animals)
{
    if (animal is Animal concreteAnimal)
    {
        concreteAnimal.DisplayInfo();
        concreteAnimal.CheckEnergy();
        Console.WriteLine();
    }
}

// Демонстрация специфических методов
Console.WriteLine("=== СПЕЦИФИЧЕСКИЕ МЕТОДЫ ===\n");

Dog dog = new Dog("Рекс", 4, 15.0, "Доберман", "Петр");
dog.Fetch();
dog.DisplayInfo();

Console.WriteLine();

Cat cat = new Cat("Мурка", 3, 3.8, "серая", false);
cat.Purr();
cat.DisplayInfo();

Console.WriteLine();

Bird bird = new Bird("Гоша", 2, 0.4, 20.0, true);
bird.BuildNest();
bird.DisplayInfo();

Console.WriteLine();

Fish fish = new Fish("Золотистый", 2, 0.6, "пресная", 1.8);
fish.Swim();
fish.DisplayInfo();

// Дополнительная демонстрация взаимодействия
Console.WriteLine("=== ВЗАИМОДЕЙСТВИЕ ЖИВОТНЫХ ===\n");

Animal[] animalArray = new Animal[]
{
    new Dog("Бобик", 2, 10.0, "Дворняжка", "Сергей"),
    new Cat("Васька", 1, 3.5, "черный", true),
    new Bird("Соловей", 1, 0.2, 12.0, true)
};

foreach (var animal in animalArray)
{
    Console.WriteLine($"--- {animal.Name} ---");
    animal.MakeSound();
    animal.Move();
    animal.Eat("любимая еда");
    animal.DisplayInfo();
    animal.CheckEnergy();
    Console.WriteLine();
}

// Демонстрация заменяемости через интерфейс
Console.WriteLine("=== ЗАМЕНЯЕМОСТЬ ЧЕРЕЗ ИНТЕРФЕЙС ===\n");
IAnimal anyAnimal = new Dog("Дружок", 5, 8.0, "Такса", "Анна");
anyAnimal.MakeSound();
anyAnimal.Move();

anyAnimal = new Cat("Снежок", 4, 5.0, "белый", true);
anyAnimal.MakeSound();
anyAnimal.Move();
````
