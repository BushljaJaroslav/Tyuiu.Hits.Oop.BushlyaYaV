```
public class Book
{
    // Приватные поля для инкапсуляции
    private string _title;
    private string _author;
    private int _yearOfPublication;
    private bool _isBorrowed;
    private Reader _currentReader;
    private int _pageCount; // Новый атрибут
    private string _publisher; // Новый атрибут
    private decimal _price; // Новый атрибут
    private int _readCount; // Новый атрибут для подсчета прочтений

    // Конструктор с использованием свойств
    public Book(string title, string author, int yearOfPublication, int pageCount, string publisher, decimal price)
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
        PageCount = pageCount;
        Publisher = publisher;
        Price = price;
        IsBorrowed = false;
        CurrentReader = null;
        ReadCount = 0;
    }

    // Свойства с геттерами и сеттерами
    public string Title
    {
        get { return _title; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _title = value;
            else
                throw new ArgumentException("Название книги не может быть пустым!");
        }
    }

    public string Author
    {
        get { return _author; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _author = value;
            else
                throw new ArgumentException("Автор не может быть пустым!");
        }
    }

    public int YearOfPublication
    {
        get { return _yearOfPublication; }
        set 
        { 
            if (value > 0 && value <= DateTime.Now.Year)
                _yearOfPublication = value;
            else
                throw new ArgumentOutOfRangeException("Год издания должен быть положительным и не больше текущего года!");
        }
    }

    public int PageCount
    {
        get { return _pageCount; }
        set 
        { 
            if (value > 0)
                _pageCount = value;
            else
                throw new ArgumentException("Количество страниц должно быть положительным!");
        }
    }

    public string Publisher
    {
        get { return _publisher; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _publisher = value;
            else
                throw new ArgumentException("Издатель не может быть пустым!");
        }
    }

    public decimal Price
    {
        get { return _price; }
        set 
        { 
            if (value >= 0)
                _price = value;
            else
                throw new ArgumentException("Цена не может быть отрицательной!");
        }
    }

    public int ReadCount
    {
        get { return _readCount; }
        private set { _readCount = value; }
    }

    public bool IsBorrowed
    {
        get { return _isBorrowed; }
        private set { _isBorrowed = value; }
    }

    public Reader CurrentReader
    {
        get { return _currentReader; }
        private set { _currentReader = value; }
    }

    // Перегрузка методов Read()
    public virtual void Read()
    {
        ReadCount++;
        Console.WriteLine($"Чтение книги: '{Title}' (страниц: {PageCount})");
    }

    public virtual void Read(int pages) // Перегруженный метод
    {
        ReadCount++;
        Console.WriteLine($"Чтение книги: '{Title}' - прочитано {pages} страниц из {PageCount}");
    }

    public virtual void Read(string readingMode) // Перегруженный метод
    {
        ReadCount++;
        Console.WriteLine($"Чтение книги: '{Title}' в режиме '{readingMode}'");
    }

    // Перегрузка методов GetInfo()
    public virtual string GetInfo()
    {
        string status = IsBorrowed ? $" (Взята читателем: {CurrentReader.Name})" : " (Доступна)";
        return $"Книга: '{Title}', Автор: {Author}, Год: {YearOfPublication}, Страниц: {PageCount}, Издатель: {Publisher}{status}";
    }

    public virtual string GetInfo(bool includePrice) // Перегруженный метод
    {
        string baseInfo = GetInfo();
        return includePrice ? baseInfo + $", Цена: {Price:C}" : baseInfo;
    }

    public virtual string GetInfo(string additionalInfo) // Перегруженный метод
    {
        return GetInfo() + $", {additionalInfo}";
    }

    // Перегрузка методов Borrow()
    public virtual void Borrow(Reader reader)
    {
        if (IsBorrowed)
        {
            Console.WriteLine($"Книга '{Title}' уже взята читателем {CurrentReader.Name}");
            return;
        }

        IsBorrowed = true;
        CurrentReader = reader;
        Console.WriteLine($"Книга '{Title}' выдана читателю {reader.Name}");
        reader.AddBorrowedBook(this);
    }

    public virtual void Borrow(Reader reader, int days) // Перегруженный метод
    {
        Borrow(reader);
        Console.WriteLine($"Срок возврата: {DateTime.Now.AddDays(days):dd.MM.yyyy}");
    }

    public void Return()
    {
        if (!IsBorrowed)
        {
            Console.WriteLine($"Книга '{Title}' уже возвращена");
            return;
        }

        Console.WriteLine($"Книга '{Title}' возвращена читателем {CurrentReader.Name}");
        IsBorrowed = false;
        CurrentReader.RemoveBorrowedBook(this);
        CurrentReader = null;
    }

    // Новые методы
    public virtual double CalculateReadingTime(double pagesPerHour)
    {
        return PageCount / pagesPerHour;
    }

    public virtual bool IsValuable()
    {
        return YearOfPublication < 1950 || Price > 1000;
    }

    public void IncrementReadCount()
    {
        ReadCount++;
    }

    public bool IsAvailable()
    {
        return !IsBorrowed;
    }
}

// Производный класс Textbook (Учебник)
public class Textbook : Book
{
    private string _subject;
    private string _educationLevel; // Новый атрибут
    private bool _hasExercises; // Новый атрибут

    public Textbook(string title, string author, int yearOfPublication, int pageCount, 
                   string publisher, decimal price, string subject, string educationLevel, bool hasExercises) 
        : base(title, author, yearOfPublication, pageCount, publisher, price)
    {
        Subject = subject;
        EducationLevel = educationLevel;
        HasExercises = hasExercises;
    }

    public string Subject
    {
        get { return _subject; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _subject = value;
            else
                throw new ArgumentException("Предмет не может быть пустым!");
        }
    }

    public string EducationLevel
    {
        get { return _educationLevel; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _educationLevel = value;
            else
                throw new ArgumentException("Уровень образования не может быть пустым!");
        }
    }

    public bool HasExercises
    {
        get { return _hasExercises; }
        set { _hasExercises = value; }
    }

    // Переопределение методов (перекрытие)
    public override void Read()
    {
        base.Read();
        Console.WriteLine($"Изучение учебника по предмету '{Subject}' (уровень: {EducationLevel})");
        if (HasExercises)
        {
            Console.WriteLine("Выполнение упражнений...");
        }
    }

    public override void Read(int pages)
    {
        base.Read(pages);
        Console.WriteLine($"Изучение учебника '{Title}' - проработано {pages} страниц");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}, Уровень: {EducationLevel}, Упражнения: {(HasExercises ? "Да" : "Нет")}";
    }

    public override void Borrow(Reader reader)
    {
        base.Borrow(reader);
        Console.WriteLine($"Учебник по предмету '{Subject}' требует внимательного изучения");
    }

    public override void Borrow(Reader reader, int days)
    {
        base.Borrow(reader, days);
        Console.WriteLine($"Рекомендуется делать конспекты при изучении");
    }

    // Переопределение новых методов
    public override double CalculateReadingTime(double pagesPerHour)
    {
        // Учебники читаются медленнее
        return base.CalculateReadingTime(pagesPerHour * 0.7);
    }

    public override bool IsValuable()
    {
        return base.IsValuable() || EducationLevel == "Высшее образование";
    }

    // Новый метод для учебника
    public void TakeTest()
    {
        Console.WriteLine($"Проверка знаний по учебнику '{Title}'...");
        Console.WriteLine("Тест завершен!");
    }
}

// Производный класс Fiction (Художественная литература)
public class Fiction : Book
{
    private string _genre;
    private string _literaryPeriod; // Новый атрибут
    private bool _isBestseller; // Новый атрибут

    public Fiction(string title, string author, int yearOfPublication, int pageCount, 
                  string publisher, decimal price, string genre, string literaryPeriod, bool isBestseller) 
        : base(title, author, yearOfPublication, pageCount, publisher, price)
    {
        Genre = genre;
        LiteraryPeriod = literaryPeriod;
        IsBestseller = isBestseller;
    }

    public string Genre
    {
        get { return _genre; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _genre = value;
            else
                throw new ArgumentException("Жанр не может быть пустым!");
        }
    }

    public string LiteraryPeriod
    {
        get { return _literaryPeriod; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _literaryPeriod = value;
            else
                throw new ArgumentException("Литературный период не может быть пустым!");
        }
    }

    public bool IsBestseller
    {
        get { return _isBestseller; }
        set { _isBestseller = value; }
    }

    // Переопределение методов (перекрытие)
    public override void Read()
    {
        base.Read();
        Console.WriteLine($"Погружение в художественное произведение жанра '{Genre}'");
        if (IsBestseller)
        {
            Console.WriteLine("Это бестселлер - наслаждайтесь чтением!");
        }
    }

    public override void Read(string readingMode)
    {
        base.Read(readingMode);
        Console.WriteLine($"Чтение художественной литературы в режиме '{readingMode}'");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Жанр: {Genre}, Период: {LiteraryPeriod}, Бестселлер: {(IsBestseller ? "Да" : "Нет")}";
    }

    public override void Borrow(Reader reader)
    {
        base.Borrow(reader);
        Console.WriteLine($"Художественная книга в жанре '{Genre}' выдана для удовольствия чтения");
    }

    // Переопределение новых методов
    public override double CalculateReadingTime(double pagesPerHour)
    {
        // Художественную литературу читают быстрее
        return base.CalculateReadingTime(pagesPerHour * 1.2);
    }

    public override bool IsValuable()
    {
        return base.IsValuable() || LiteraryPeriod == "Классика";
    }

    // Новые методы для художественной литературы
    public void AnalyzeCharacters()
    {
        Console.WriteLine($"Анализ персонажей книги '{Title}'...");
        Console.WriteLine("Анализ завершен!");
    }

    public void RecommendSimilarBooks(Fiction[] books)
    {
        Console.WriteLine($"Похожие книги в жанре '{Genre}':");
        foreach (var book in books)
        {
            if (book.Genre == Genre && book.Title != Title)
            {
                Console.WriteLine($"- {book.Title} ({book.Author})");
            }
        }
    }
}

// Производный класс ScientificLiterature (Научная литература)
public class ScientificLiterature : Book
{
    private string _fieldOfScience;
    private string _researchArea; // Новый атрибут
    private int _citationCount; // Новый атрибут

    public ScientificLiterature(string title, string author, int yearOfPublication, int pageCount, 
                               string publisher, decimal price, string fieldOfScience, string researchArea, int citationCount) 
        : base(title, author, yearOfPublication, pageCount, publisher, price)
    {
        FieldOfScience = fieldOfScience;
        ResearchArea = researchArea;
        CitationCount = citationCount;
    }

    public string FieldOfScience
    {
        get { return _fieldOfScience; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _fieldOfScience = value;
            else
                throw new ArgumentException("Область науки не может быть пустой!");
        }
    }

    public string ResearchArea
    {
        get { return _researchArea; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _researchArea = value;
            else
                throw new ArgumentException("Область исследования не может быть пустой!");
        }
    }

    public int CitationCount
    {
        get { return _citationCount; }
        set 
        { 
            if (value >= 0)
                _citationCount = value;
            else
                throw new ArgumentException("Количество цитирований не может быть отрицательным!");
        }
    }

    // Переопределение методов (перекрытие)
    public override void Read()
    {
        base.Read();
        Console.WriteLine($"Исследование научной работы в области '{FieldOfScience}'");
        Console.WriteLine($"Область исследования: {ResearchArea}");
    }

    public override void Read(int pages)
    {
        base.Read(pages);
        Console.WriteLine($"Изучение научной литературы '{Title}' - проанализировано {pages} страниц");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Область науки: {FieldOfScience}, Исследования: {ResearchArea}, Цитирований: {CitationCount}";
    }

    public override void Borrow(Reader reader)
    {
        base.Borrow(reader);
        Console.WriteLine($"Научная литература требует специальных знаний в области '{FieldOfScience}'");
    }

    public override void Borrow(Reader reader, int days)
    {
        base.Borrow(reader, days);
        Console.WriteLine($"Рекомендуется вести исследовательские заметки");
    }

    // Переопределение новых методов
    public override double CalculateReadingTime(double pagesPerHour)
    {
        // Научную литературу читают медленнее
        return base.CalculateReadingTime(pagesPerHour * 0.5);
    }

    public override bool IsValuable()
    {
        return base.IsValuable() || CitationCount > 100;
    }

    // Новые методы для научной литературы
    public void Cite()
    {
        CitationCount++;
        Console.WriteLine($"Работа '{Title}' процитирована. Всего цитирований: {CitationCount}");
    }

    public void Research()
    {
        Console.WriteLine($"Проведение исследования на основе работы '{Title}'...");
        Console.WriteLine("Исследование завершено!");
    }
}

// Класс для читателя с дополнительными атрибутами
public class Reader
{
    private string _name;
    private string _email;
    private string _phone; // Новый атрибут
    private DateTime _registrationDate; // Новый атрибут
    private Book[] _borrowedBooks;
    private int _bookCount;

    public Reader(string name, string email, string phone)
    {
        Name = name;
        Email = email;
        Phone = phone;
        RegistrationDate = DateTime.Now;
        _borrowedBooks = new Book[10];
        _bookCount = 0;
    }

    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _name = value;
            else
                throw new ArgumentException("Имя читателя не может быть пустым!");
        }
    }

    public string Email
    {
        get { return _email; }
        set 
        { 
            if (!string.IsNullOrEmpty(value) && value.Contains("@"))
                _email = value;
            else
                throw new ArgumentException("Email должен быть корректным!");
        }
    }

    public string Phone
    {
        get { return _phone; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                _phone = value;
            else
                throw new ArgumentException("Телефон не может быть пустым!");
        }
    }

    public DateTime RegistrationDate
    {
        get { return _registrationDate; }
        private set { _registrationDate = value; }
    }

    public void AddBorrowedBook(Book book)
    {
        if (_bookCount < _borrowedBooks.Length)
        {
            _borrowedBooks[_bookCount] = book;
            _bookCount++;
        }
        else
        {
            Console.WriteLine($"{Name} уже взял максимальное количество книг");
        }
    }

    public void RemoveBorrowedBook(Book book)
    {
        for (int i = 0; i < _bookCount; i++)
        {
            if (_borrowedBooks[i] == book)
            {
                for (int j = i; j < _bookCount - 1; j++)
                {
                    _borrowedBooks[j] = _borrowedBooks[j + 1];
                }
                _bookCount--;
                break;
            }
        }
    }

    public void DisplayBorrowedBooks()
    {
        Console.WriteLine($"\nКниги, взятые {Name}:");
        if (_bookCount == 0)
        {
            Console.WriteLine("Нет взятых книг");
            return;
        }

        for (int i = 0; i < _bookCount; i++)
        {
            Console.WriteLine($"- {_borrowedBooks[i].Title} ({_borrowedBooks[i].GetType().Name})");
        }
    }

    public Book FindBorrowedBook(string title)
    {
        for (int i = 0; i < _bookCount; i++)
        {
            if (_borrowedBooks[i].Title == title)
            {
                return _borrowedBooks[i];
            }
        }
        return null;
    }

    // Новые методы
    public int GetReadingExperience()
    {
        return (DateTime.Now - RegistrationDate).Days;
    }

    public bool CanBorrowMoreBooks()
    {
        return _bookCount < _borrowedBooks.Length;
    }
}

// Generic класс для управления коллекциями книг
public class LibraryCollection<T> where T : Book
{
    private List<T> _books;
    public string CollectionName { get; set; }
    public string Category { get; set; }

    public LibraryCollection(string name, string category)
    {
        CollectionName = name;
        Category = category;
        _books = new List<T>();
    }

    public void AddBook(T book)
    {
        _books.Add(book);
        Console.WriteLine($"Добавлена книга '{book.Title}' в коллекцию '{CollectionName}'");
    }

    public void RemoveBook(T book)
    {
        if (_books.Remove(book))
        {
            Console.WriteLine($"Удалена книга '{book.Title}' из коллекции '{CollectionName}'");
        }
    }

    public void DisplayCollection()
    {
        Console.WriteLine($"\n--- Коллекция '{CollectionName}' ({Category}) ---");
        if (_books.Count == 0)
        {
            Console.WriteLine("Коллекция пуста");
            return;
        }

        foreach (var book in _books)
        {
            Console.WriteLine($"- {book.GetInfo()}");
        }
    }

    public T FindBookByTitle(string title)
    {
        return _books.FirstOrDefault(b => b.Title.Equals(title, StringComparison.OrdinalIgnoreCase));
    }

    public List<T> FindBooksByAuthor(string author)
    {
        return _books.Where(b => b.Author.Equals(author, StringComparison.OrdinalIgnoreCase)).ToList();
    }

    public void ReadAllBooks()
    {
        Console.WriteLine($"\n--- Чтение всех книг в коллекции '{CollectionName}' ---");
        foreach (var book in _books)
        {
            book.Read();
        }
    }

    public double CalculateTotalValue()
    {
        return _books.Sum(b => (double)b.Price);
    }

    public int BookCount => _books.Count;
}

// Generic класс для системы рекомендаций
public class RecommendationSystem<T> where T : Book
{
    private List<T> _allBooks;
    private Dictionary<string, List<T>> _userPreferences;

    public RecommendationSystem()
    {
        _allBooks = new List<T>();
        _userPreferences = new Dictionary<string, List<T>>();
    }

    public void AddBook(T book)
    {
        _allBooks.Add(book);
    }

    public void AddUserPreference(string userName, T book)
    {
        if (!_userPreferences.ContainsKey(userName))
        {
            _userPreferences[userName] = new List<T>();
        }
        _userPreferences[userName].Add(book);
    }

    public List<T> GetRecommendations(string userName)
    {
        if (!_userPreferences.ContainsKey(userName) || _userPreferences[userName].Count == 0)
        {
            return _allBooks.Take(3).ToList(); // Возвращаем первые 3 книги, если нет предпочтений
        }

        var userBooks = _userPreferences[userName];
        var recommendations = new List<T>();

        // Простая логика рекомендаций на основе жанров/предметов
        foreach (var userBook in userBooks)
        {
            var similarBooks = _allBooks.Where(b => 
                b != userBook && 
                (b is Fiction fiction && userBook is Fiction userFiction && fiction.Genre == userFiction.Genre) ||
                (b is Textbook textbook && userBook is Textbook userTextbook && textbook.Subject == userTextbook.Subject) ||
                (b is ScientificLiterature sciLit && userBook is ScientificLiterature userSciLit && sciLit.FieldOfScience == userSciLit.FieldOfScience)
            ).Take(2);

            recommendations.AddRange(similarBooks);
        }

        return recommendations.Distinct().Take(5).ToList();
    }

    public void DisplayRecommendations(string userName)
    {
        var recommendations = GetRecommendations(userName);
        Console.WriteLine($"\n--- Рекомендации для {userName} ---");
        
        if (recommendations.Count == 0)
        {
            Console.WriteLine("Нет рекомендаций");
            return;
        }

        foreach (var book in recommendations)
        {
            Console.WriteLine($"- {book.Title} ({book.Author})");
        }
    }
}

// Класс библиотеки с использованием Generic классов
public class AdvancedLibrary
{
    private LibraryCollection<Book> _allBooks;
    private LibraryCollection<Textbook> _textbooks;
    private LibraryCollection<Fiction> _fiction;
    private LibraryCollection<ScientificLiterature> _scientificLiterature;
    private Reader[] _readers;
    private int _readerCount;
    private RecommendationSystem<Book> _recommendationSystem;

    public AdvancedLibrary()
    {
        _allBooks = new LibraryCollection<Book>("Все книги", "Универсальная");
        _textbooks = new LibraryCollection<Textbook>("Учебники", "Образование");
        _fiction = new LibraryCollection<Fiction>("Художественная литература", "Развлечения");
        _scientificLiterature = new LibraryCollection<ScientificLiterature>("Научная литература", "Наука");
        _readers = new Reader[50];
        _readerCount = 0;
        _recommendationSystem = new RecommendationSystem<Book>();
    }

    public void AddBook(Book book)
    {
        _allBooks.AddBook(book);
        _recommendationSystem.AddBook(book);

        // Добавление в специализированные коллекции
        if (book is Textbook textbook)
            _textbooks.AddBook(textbook);
        else if (book is Fiction fiction)
            _fiction.AddBook(fiction);
        else if (book is ScientificLiterature sciLit)
            _scientificLiterature.AddBook(sciLit);
    }

    public void RegisterReader(Reader reader)
    {
        if (_readerCount < _readers.Length)
        {
            _readers[_readerCount] = reader;
            _readerCount++;
        }
    }

    public void DisplayAllCollections()
    {
        _allBooks.DisplayCollection();
        _textbooks.DisplayCollection();
        _fiction.DisplayCollection();
        _scientificLiterature.DisplayCollection();
    }

    public void DisplayRecommendationsForUser(string userName)
    {
        _recommendationSystem.DisplayRecommendations(userName);
    }

    public void AddUserPreference(string userName, Book book)
    {
        _recommendationSystem.AddUserPreference(userName, book);
    }

    public void DemonstratePolymorphism()
    {
        Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ===");
        
        Book[] books = new Book[]
        {
            new Textbook("Математика", "Автор1", 2020, 300, "Изд1", 500, "Математика", "Высшее", true),
            new Fiction("Роман", "Автор2", 2019, 250, "Изд2", 300, "Драма", "Современная", false),
            new ScientificLiterature("Физика", "Автор3", 2021, 400, "Изд3", 800, "Физика", "Квантовая", 50)
        };

        // Демонстрация переопределения методов
        foreach (var book in books)
        {
            Console.WriteLine($"\n--- {book.GetType().Name} ---");
            book.Read(); // Полиморфный вызов
            book.Read(50); // Перегруженный метод
            Console.WriteLine($"Ценная: {book.IsValuable()}");
            Console.WriteLine($"Время чтения: {book.CalculateReadingTime(30):F1} часов");
        }

        // Демонстрация перегрузки методов GetInfo
        Console.WriteLine("\n--- Перегрузка методов GetInfo ---");
        foreach (var book in books)
        {
            Console.WriteLine(book.GetInfo());
            Console.WriteLine(book.GetInfo(true)); // С ценой
            Console.WriteLine(book.GetInfo("Дополнительная информация"));
            Console.WriteLine();
        }
    }
}

// Демонстрация работы расширенной системы
Console.WriteLine("=== РАСШИРЕННАЯ БИБЛИОТЕЧНАЯ СИСТЕМА ===\n");

AdvancedLibrary library = new AdvancedLibrary();

// Создание книг с дополнительными атрибутами
Textbook mathTextbook = new Textbook("Высшая математика", "Иванов", 2020, 400, "Наука", 1200, "Математика", "Высшее образование", true);
Fiction novel = new Fiction("Война и мир", "Толстой", 1869, 1225, "Классика", 800, "Роман", "Классика", true);
ScientificLiterature physicsBook = new ScientificLiterature("Квантовая физика", "Фейнман", 1965, 600, "MIT", 1500, "Физика", "Квантовая механика", 5000);
Fiction dystopia = new Fiction("О дивный новый мир", "Хаксли", 1932, 350, "Penguin", 450, "Антиутопия", "Модернизм", true);

// Добавление книг в библиотеку
library.AddBook(mathTextbook);
library.AddBook(novel);
library.AddBook(physicsBook);
library.AddBook(dystopia);

// Создание читателей
Reader reader1 = new Reader("Алексей", "alex@mail.com", "+79123456789");
Reader reader2 = new Reader("Мария", "maria@mail.com", "+79098765432");

library.RegisterReader(reader1);
library.RegisterReader(reader2);

// Отображение всех коллекций
library.DisplayAllCollections();

// Демонстрация полиморфизма
library.DemonstratePolymorphism();

// Работа с рекомендательной системой
library.AddUserPreference("Алексей", mathTextbook);
library.AddUserPreference("Алексей", physicsBook);
library.AddUserPreference("Мария", novel);
library.AddUserPreference("Мария", dystopia);

library.DisplayRecommendationsForUser("Алексей");
library.DisplayRecommendationsForUser("Мария");

// Демонстрация работы с книгами
Console.WriteLine("\n=== РАБОТА С КНИГАМИ ===");
mathTextbook.Borrow(reader1, 14); // Перегруженный метод
novel.Read("вечерний режим"); // Перегруженный метод
physicsBook.Cite(); // Специфичный метод
mathTextbook.TakeTest(); // Специфичный метод

// Демонстрация валидации свойств
Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ВАЛИДАЦИИ ===");
try
{
    Book invalidBook = new Book("", "Автор", 2020, 100, "Издатель", 500);
}
catch (ArgumentException ex)
{
    Console.WriteLine($"Ошибка: {ex.Message}");
}

try
{
    Textbook invalidTextbook = new Textbook("Учебник", "", 2020, 100, "Издатель", 500, "Предмет", "Уровень", true);
}
catch (ArgumentException ex)
{
    Console.WriteLine($"Ошибка: {ex.Message}");
}

Console.WriteLine("\n=== СТАТИСТИКА ===");
Console.WriteLine($"Всего книг в библиотеке: 4");
Console.WriteLine($"Учебников: 1");
Console.WriteLine($"Художественной литературы: 2");
Console.WriteLine($"Научной литературы: 1");
Console.WriteLine($"Зарегистрированных читателей: 2");
```
