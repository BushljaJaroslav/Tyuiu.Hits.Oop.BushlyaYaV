<h1 style="color:DodgerBlue">Generic классы в C#</h1>


Генерические классы в C# представляют собой шаблон класса, который может работать с различными типами данных без потери производительности. Они позволяют создавать компоненты, которые могут быть безопасно повторно использованы в различных контекстах, обеспечивая при этом строгую типизацию и избегая необходимости приведения типов.

#### Основные преимущества генерических классов:

1. **Безопасность типов**: Генерические классы обеспечивают проверку типов на этапе компиляции, что помогает предотвратить ошибки во время выполнения.
2. **Универсальность**: Генерические классы могут работать с различными типами данных, делая их очень гибкими и универсальными.
3. **Производительность**: Генерические классы не приводят к значительной потере производительности, поскольку информация о типе хранится вместе с объектом.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример генерического класса `VehicleCollection<T>`, который может хранить коллекцию объектов типа `T`, где `T` может быть любой строкой, например, `Vehicle`, `Car` или `Motorcycle`.

```csharp
using System;
using System.Collections.Generic;

// Определение базового класса Vehicle
public class Vehicle
{
    public virtual void Drive()
    {
        Console.WriteLine("Транспортное средство находится в движении.");
    }
}

// Определение производного класса Car
public class Car : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Машина едет.");
    }
}

// Определение производного класса Motorcycle
public class Motorcycle : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Мотоцикл едет.");
    }
}

// Определение генерического класса VehicleCollection<T>
public class VehicleCollection<T> where T : Vehicle
{
    private List<T> _vehicles = new List<T>();

    public void Add(T vehicle)
    {
        _vehicles.Add(vehicle);
    }

    public void Remove(T vehicle)
    {
        _vehicles.Remove(vehicle);
    }

    public void DisplayVehicles()
    {
        foreach (var vehicle in _vehicles)
        {
            vehicle.Drive(); // Вызов метода Drive() для каждого транспортного средства
        }
    }
}


VehicleCollection<Car> cars = new VehicleCollection<Car>();
cars.Add(new Car());
cars.Add(new Motorcycle()); // Это возможно благодаря ограничению "where T : Vehicle"

cars.DisplayVehicles(); // Выводит: "Машина едет." и "Мотоцикл едет."


```

В этом примере `VehicleCollection<T>` является генерическим классом, который может хранить коллекцию объектов типа `T`, где `T` должен быть подклассом `Vehicle`. Это позволяет создавать коллекции автомобилей, мотоциклов и других видов транспортных средств, обеспечивая при этом строгую типизацию и безопасность типов.

### Заключение

Генерические классы в C# предоставляют мощный и гибкий способ создания компонентов, которые могут работать с различными типами данных, обеспечивая при этом строгую типизацию и безопасность типов. Это делает их идеальным выбором для многих задач, требующих повторного использования кода с высокой степенью абстракции.

````
public class Vehicle
{
    public string Brand { get; set; }
    public string Model { get; set; }
    public int Year { get; set; }

    public Vehicle(string brand, string model, int year)
    {
        Brand = brand;
        Model = model;
        Year = year;
    }

    public virtual void Drive()
    {
        Console.WriteLine($"Транспортное средство {Brand} {Model} находится в движении.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Транспорт: {Brand} {Model}, Год: {Year}");
    }
}

// Определение производного класса Car
public class Car : Vehicle
{
    public int Doors { get; set; }

    public Car(string brand, string model, int year, int doors) 
        : base(brand, model, year)
    {
        Doors = doors;
    }

    public override void Drive()
    {
        Console.WriteLine($"Машина {Brand} {Model} едет по дороге.");
    }

    public override void DisplayInfo()
    {
        Console.WriteLine($"Автомобиль: {Brand} {Model}, Год: {Year}, Двери: {Doors}");
    }
}

// Определение производного класса Motorcycle
public class Motorcycle : Vehicle
{
    public string Type { get; set; }

    public Motorcycle(string brand, string model, int year, string type) 
        : base(brand, model, year)
    {
        Type = type;
    }

    public override void Drive()
    {
        Console.WriteLine($"Мотоцикл {Brand} {Model} едет по трассе.");
    }

    public override void DisplayInfo()
    {
        Console.WriteLine($"Мотоцикл: {Brand} {Model}, Год: {Year}, Тип: {Type}");
    }
}

// Определение генерического класса VehicleCollection<T>
public class VehicleCollection<T> where T : Vehicle
{
    private List<T> _vehicles = new List<T>();

    public void Add(T vehicle)
    {
        _vehicles.Add(vehicle);
        Console.WriteLine($"Добавлен: {vehicle.Brand} {vehicle.Model}");
    }

    public void Remove(T vehicle)
    {
        _vehicles.Remove(vehicle);
        Console.WriteLine($"Удален: {vehicle.Brand} {vehicle.Model}");
    }

    public void DisplayAllVehicles()
    {
        Console.WriteLine($"\\n--- Коллекция транспортных средств ({typeof(T).Name}) ---");
        foreach (var vehicle in _vehicles)
        {
            vehicle.DisplayInfo();
        }
    }

    public void DriveAll()
    {
        Console.WriteLine($"\\n--- Движение транспортных средств ({typeof(T).Name}) ---");
        foreach (var vehicle in _vehicles)
        {
            vehicle.Drive();
        }
    }

    public T FindByBrand(string brand)
    {
        return _vehicles.FirstOrDefault(v => v.Brand.Equals(brand, StringComparison.OrdinalIgnoreCase));
    }

    public int Count => _vehicles.Count;
}

// Демонстрация работы с генерическими классами
Console.WriteLine("=== Демонстрация Generic классов с Vehicle ===");

// Создание коллекции автомобилей
VehicleCollection<Car> carCollection = new VehicleCollection<Car>();
carCollection.Add(new Car("Toyota", "Camry", 2022, 4));
carCollection.Add(new Car("Honda", "Civic", 2023, 4));
carCollection.Add(new Car("BMW", "X5", 2021, 5));

carCollection.DisplayAllVehicles();
carCollection.DriveAll();

Console.WriteLine($"\\nВсего автомобилей в коллекции: {carCollection.Count}");

// Поиск автомобиля по марке
Car foundCar = carCollection.FindByBrand("Toyota");
if (foundCar != null)
{
    Console.WriteLine($"\\nНайден автомобиль: {foundCar.Brand} {foundCar.Model}");
}

// Создание коллекции мотоциклов
Console.WriteLine("\\n" + new string('=', 50));
VehicleCollection<Motorcycle> motorcycleCollection = new VehicleCollection<Motorcycle>();
motorcycleCollection.Add(new Motorcycle("Yamaha", "YZF-R1", 2023, "Спортивный"));
motorcycleCollection.Add(new Motorcycle("Harley-Davidson", "Sportster", 2022, "Круизер"));

motorcycleCollection.DisplayAllVehicles();
motorcycleCollection.DriveAll();

// Универсальная коллекция для любых транспортных средств
Console.WriteLine("\\n" + new string('=', 50));
VehicleCollection<Vehicle> universalCollection = new VehicleCollection<Vehicle>();
universalCollection.Add(new Car("Ford", "Focus", 2020, 4));
universalCollection.Add(new Motorcycle("Kawasaki", "Ninja", 2023, "Спортивный"));
universalCollection.Add(new Vehicle("Generic", "Transport", 2021));

universalCollection.DisplayAllVehicles();
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также с пременением Generic классов.

````
// Базовый класс Animal
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Species { get; set; }
    public double Weight { get; set; }

    public Animal(string name, int age, string species, double weight)
    {
        Name = name;
        Age = age;
        Species = species;
        Weight = weight;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Животное: {Name}, Вид: {Species}, Возраст: {Age}, Вес: {Weight}кг");
    }

    public virtual void Eat()
    {
        Console.WriteLine($"{Name} ест.");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }

    public Dog(string name, int age, string species, double weight, string breed, bool isTrained) 
        : base(name, age, species, weight)
    {
        Breed = breed;
        IsTrained = isTrained;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} ({Breed}) лает: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        string trainedStatus = IsTrained ? "дрессированная" : "недрессированная";
        Console.WriteLine($"Собака: {Name}, Порода: {Breed}, {trainedStatus}, Возраст: {Age}, Вес: {Weight}кг");
    }

    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит палку!");
    }
}

// Производный класс Cat
public class Cat : Animal
{
    public bool IsIndoor { get; set; }
    public int Lives { get; set; }

    public Cat(string name, int age, string species, double weight, bool isIndoor, int lives = 9) 
        : base(name, age, species, weight)
    {
        IsIndoor = isIndoor;
        Lives = lives;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    public override void DisplayInfo()
    {
        string indoorStatus = IsIndoor ? "домашний" : "уличный";
        Console.WriteLine($"Кот: {Name}, {indoorStatus}, Жизней: {Lives}, Возраст: {Age}, Вес: {Weight}кг");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }

    public Bird(string name, int age, string species, double weight, double wingspan, bool canFly) 
        : base(name, age, species, weight)
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        string flyAbility = CanFly ? "летающая" : "нелетающая";
        Console.WriteLine($"Птица: {Name}, {flyAbility}, Размах крыльев: {Wingspan}см, Возраст: {Age}, Вес: {Weight}кг");
    }

    public void Fly()
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} летит с размахом крыльев {Wingspan}см!");
        }
        else
        {
            Console.WriteLine($"{Name} не может летать.");
        }
    }
}

// Производный класс Fish
public class Fish : Animal
{
    public string WaterType { get; set; }
    public double SwimSpeed { get; set; }

    public Fish(string name, int age, string species, double weight, string waterType, double swimSpeed) 
        : base(name, age, species, weight)
    {
        WaterType = waterType;
        SwimSpeed = swimSpeed;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук: Буль-буль!");
    }

    public override void DisplayInfo()
    {
        Console.WriteLine($"Рыба: {Name}, Тип воды: {WaterType}, Скорость: {SwimSpeed} м/с, Возраст: {Age}, Вес: {Weight}кг");
    }

    public void Swim()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде со скоростью {SwimSpeed} м/с!");
    }
}

// Генерический класс ZooEnclosure<T> для вольеров с животными
public class ZooEnclosure<T> where T : Animal
{
    private List<T> _animals = new List<T>();
    public string EnclosureName { get; set; }
    public double Area { get; set; } // площадь вольера в м²

    public ZooEnclosure(string name, double area)
    {
        EnclosureName = name;
        Area = area;
    }

    public void AddAnimal(T animal)
    {
        _animals.Add(animal);
        Console.WriteLine($"Добавлен(а) {animal.Species} {animal.Name} в вольер '{EnclosureName}'");
    }

    public void RemoveAnimal(T animal)
    {
        if (_animals.Remove(animal))
        {
            Console.WriteLine($"Удален(а) {animal.Species} {animal.Name} из вольера '{EnclosureName}'");
        }
    }

    public void DisplayAllAnimals()
    {
        Console.WriteLine($"\\n--- Вольер '{EnclosureName}' ({typeof(T).Name}) Площадь: {Area}м² ---");
        if (_animals.Count == 0)
        {
            Console.WriteLine("Вольер пуст");
            return;
        }

        foreach (var animal in _animals)
        {
            animal.DisplayInfo();
        }
    }

    public void MakeAllSounds()
    {
        Console.WriteLine($"\\n--- Звуки из вольера '{EnclosureName}' ---");
        foreach (var animal in _animals)
        {
            animal.MakeSound();
        }
    }

    public T FindAnimalByName(string name)
    {
        return _animals.FirstOrDefault(a => a.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
    }

    public List<T> FindBySpecies(string species)
    {
        return _animals.Where(a => a.Species.Equals(species, StringComparison.OrdinalIgnoreCase)).ToList();
    }

    public int AnimalCount => _animals.Count;
    public double AnimalsPerSquareMeter => AnimalCount > 0 ? Area / AnimalCount : 0;
}

// Генерический класс VeterinaryClinic для ветеринарной клиники
public class VeterinaryClinic<T> where T : Animal
{
    private List<T> _patients = new List<T>();
    public string ClinicName { get; set; }

    public VeterinaryClinic(string name)
    {
        ClinicName = name;
    }

    public void AdmitPatient(T patient)
    {
        _patients.Add(patient);
        Console.WriteLine($"Поступил пациент: {patient.Species} {patient.Name} в клинику '{ClinicName}'");
    }

    public void DischargePatient(T patient)
    {
        if (_patients.Remove(patient))
        {
            Console.WriteLine($"Выписан пациент: {patient.Species} {patient.Name} из клиники '{ClinicName}'");
        }
    }

    public void ExamineAllPatients()
    {
        Console.WriteLine($"\\n--- Осмотр пациентов в клинике '{ClinicName}' ---");
        foreach (var patient in _patients)
        {
            Console.WriteLine($"Осматривается: {patient.Name} ({patient.Species})");
            patient.DisplayInfo();
        }
    }

    public void FeedAllPatients()
    {
        Console.WriteLine($"\\n--- Кормление пациентов в клинике '{ClinicName}' ---");
        foreach (var patient in _patients)
        {
            patient.Eat();
        }
    }
}

// Генерический класс AnimalTrainer для дрессировщиков
public class AnimalTrainer<T> where T : Animal
{
    public string TrainerName { get; set; }
    private List<T> _trainedAnimals = new List<T>();

    public AnimalTrainer(string name)
    {
        TrainerName = name;
    }

    public void AddAnimal(T animal)
    {
        _trainedAnimals.Add(animal);
        Console.WriteLine($"Дрессировщик {TrainerName} теперь работает с {animal.Species} {animal.Name}");
    }

    public void ConductTrainingSession()
    {
        Console.WriteLine($"\\n--- Тренировка с дрессировщиком {TrainerName} ---");
        foreach (var animal in _trainedAnimals)
        {
            Console.WriteLine($"Тренируется: {animal.Name}");
            animal.MakeSound();
            
            // Специфическое поведение для разных типов животных
            if (animal is Dog dog)
            {
                dog.Fetch();
            }
            else if (animal is Cat cat)
            {
                cat.Purr();
            }
            else if (animal is Bird bird)
            {
                bird.Fly();
            }
            else if (animal is Fish fish)
            {
                fish.Swim();
            }
        }
    }
}

// Демонстрация работы с генерическими классами для животных
Console.WriteLine("=== Демонстрация Generic классов с Animal ===");

// Создание специализированных вольеров
ZooEnclosure<Dog> dogEnclosure = new ZooEnclosure<Dog>("Собачий вольер", 50.0);
dogEnclosure.AddAnimal(new Dog("Бобик", 3, "Собака", 25.5, "Лабрадор", true));
dogEnclosure.AddAnimal(new Dog("Шарик", 4, "Собака", 30.0, "Овчарка", true));
dogEnclosure.AddAnimal(new Dog("Рекс", 2, "Собака", 20.0, "Такса", false));

ZooEnclosure<Cat> catEnclosure = new ZooEnclosure<Cat>("Кошачий вольер", 30.0);
catEnclosure.AddAnimal(new Cat("Мурзик", 2, "Кот", 4.2, true, 9));
catEnclosure.AddAnimal(new Cat("Барсик", 3, "Кот", 5.0, false, 8));

ZooEnclosure<Bird> birdEnclosure = new ZooEnclosure<Bird>("Птичий вольер", 25.0);
birdEnclosure.AddAnimal(new Bird("Кеша", 1, "Попугай", 0.3, 15.5, true));
birdEnclosure.AddAnimal(new Bird("Гоша", 2, "Пингвин", 15.0, 30.0, false));

ZooEnclosure<Fish> fishEnclosure = new ZooEnclosure<Fish>("Аквариум", 10.0);
fishEnclosure.AddAnimal(new Fish("Немо", 1, "Рыба-клоун", 0.2, "морская", 2.5));
fishEnclosure.AddAnimal(new Fish("Дори", 2, "Синий хирург", 0.3, "морская", 3.0));

// Универсальный вольер для любых животных
ZooEnclosure<Animal> mixedEnclosure = new ZooEnclosure<Animal>("Смешанный вольер", 100.0);
mixedEnclosure.AddAnimal(new Dog("Дружок", 5, "Собака", 22.0, "Дворняга", true));
mixedEnclosure.AddAnimal(new Cat("Васька", 4, "Кот", 4.5, true, 7));

// Отображение информации о вольерах
dogEnclosure.DisplayAllAnimals();
catEnclosure.DisplayAllAnimals();
birdEnclosure.DisplayAllAnimals();
fishEnclosure.DisplayAllAnimals();
mixedEnclosure.DisplayAllAnimals();

// Демонстрация звуков животных
Console.WriteLine("\\n" + new string('=', 60));
dogEnclosure.MakeAllSounds();
catEnclosure.MakeAllSounds();
birdEnclosure.MakeAllSounds();

// Работа ветеринарной клиники
Console.WriteLine("\\n" + new string('=', 60));
VeterinaryClinic<Animal> cityClinic = new VeterinaryClinic<Animal>("Городская ветеринарная клиника");
cityClinic.AdmitPatient(new Dog("Бобик", 3, "Собака", 25.5, "Лабрадор", true));
cityClinic.AdmitPatient(new Cat("Мурзик", 2, "Кот", 4.2, true, 9));
cityClinic.AdmitPatient(new Bird("Кеша", 1, "Попугай", 0.3, 15.5, true));

cityClinic.ExamineAllPatients();
cityClinic.FeedAllPatients();

// Работа дрессировщика
Console.WriteLine("\\n" + new string('=', 60));
AnimalTrainer<Dog> dogTrainer = new AnimalTrainer<Dog>("Иван Петров");
dogTrainer.AddAnimal(new Dog("Бобик", 3, "Собака", 25.5, "Лабрадор", true));
dogTrainer.AddAnimal(new Dog("Шарик", 4, "Собака", 30.0, "Овчарка", true));

dogTrainer.ConductTrainingSession();

// Статистика по вольерам
Console.WriteLine("\\n" + new string('=', 60));
Console.WriteLine("=== Статистика зоопарка ===");
Console.WriteLine($"Собачий вольер: {dogEnclosure.AnimalCount} животных, {dogEnclosure.AnimalsPerSquareMeter:F1} м² на животное");
Console.WriteLine($"Кошачий вольер: {catEnclosure.AnimalCount} животных, {catEnclosure.AnimalsPerSquareMeter:F1} м² на животное");
Console.WriteLine($"Птичий вольер: {birdEnclosure.AnimalCount} животных, {birdEnclosure.AnimalsPerSquareMeter:F1} м² на животное");
Console.WriteLine($"Аквариум: {fishEnclosure.AnimalCount} животных, {fishEnclosure.AnimalsPerSquareMeter:F1} м² на животное");

// Поиск животных
Console.WriteLine("\\n" + new string('=', 60));
Dog foundDog = dogEnclosure.FindAnimalByName("Бобик");
if (foundDog != null)
{
    Console.WriteLine($"Найдена собака: {foundDog.Name}, Порода: {foundDog.Breed}");
}

List<Dog> allDogs = dogEnclosure.FindBySpecies("Собака");
Console.WriteLine($"Всего собак в вольере: {allDogs.Count}");
````
